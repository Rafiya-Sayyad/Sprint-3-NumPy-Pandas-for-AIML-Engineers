## Topic 1 – loc

### Step 1 – Understand the Concept
loc selects rows and columns using their labels (the names). The format is df.loc[rows, columns].

- A single label gives one row (as a Series), and a list of labels gives many rows.
- A slice like loc[2:5] includes both the start and the end label, so the last label is included.
- A column name or a list of column names can be given as the second part.
- A colon : means all rows or all columns.
- loc can also be used to change values.

If the index is the default numbers, the labels are the row numbers 0, 1, 2 and so on. If another column is set as the index (like Name), the names are the labels.

### Step 2 – Demonstrate the Concept

#### Business Example
A manager wants the income of the customer named Ravi. If the table has Name as the index, loc["Ravi", "Income"] gives the answer directly, without searching the rows.

### Step 3 – Implement the Concept

```python
# With the default index (labels are 0, 1, 2 ...)
print(customers.loc[0])

print()
print(customers.loc[[0, 2, 4]])

# The end label is included in a slice
print()
print(customers.loc[2:5])

# Rows and columns together
print()
print(customers.loc[0:2, ["Name", "Age"]])

# A single value
print()
print("Name at label 3:", customers.loc[3, "Name"])
```

```python
# With Name as the index
named = customers.set_index("Name")

print(named.loc["Ravi"])
print()
print("Ravi's income:", named.loc["Ravi", "Income"])

# A slice of names (both ends included)
print()
print(named.loc["Asha":"Sana", ["City", "Income"]])
```

```python
# Changing a value with loc (on a copy, so the original stays the same)
temp = customers.copy()
temp.loc[0, "Income"] = 42
print("New income of row 0:", temp.loc[0, "Income"])
print("Original income    :", customers.loc[0, "Income"])
```

#### What does this program do?
In the first cell, loc[0] gives the whole first row as a Series, and loc[[0, 2, 4]] gives the rows with labels 0, 2 and 4. loc[2:5] gives the rows 2, 3, 4 and 5, which is 4 rows, because the end label is included. loc[0:2, ["Name", "Age"]] gives 3 rows and only the Name and Age columns, and loc[3, "Name"] gives a single value, Kiran.

In the second cell I made Name the index with set_index(). Now loc["Ravi"] gives Ravi's row and loc["Ravi", "Income"] gives 65.0. The name slice loc["Asha":"Sana"] gives 5 rows (Asha, Ravi, Meena, Kiran and Sana), again with both ends included, and only the City and Income columns.

In the third cell I made a copy of the table and changed the income of row 0 to 42 with loc. The copy shows 42.0 (it is a decimal because the Income column is float64), and the original still shows 40.0.

#### AI/ML Example
loc is used to pick the feature columns and the target column by name, for example X = df.loc[:, ["Age", "Income", "Purchases"]] and y = df.loc[:, "Segment"]. It is also used together with a condition to select rows and columns at the same time, and to fix wrong values in the data.

#### Advantages
- Code is easy to read because columns and rows are chosen by their names.
- Can select rows, columns, ranges and single values with the same format.
- Can also be used to change values safely.

#### Limitations
- A wrong label gives a KeyError.
- The slice includes the end label, which is different from normal Python slicing, so it can confuse beginners.
- If the index has duplicate labels, loc gives more than one row.

#### Key Point
loc selects by label, and the end of a slice is included. Format: df.loc[rows, columns].

## Topic 2 – iloc

### Step 1 – Understand the Concept
iloc selects rows and columns using their position numbers, which always start from 0, like a Python list. The format is df.iloc[rows, columns].

- A single number gives one row, and a list of numbers gives many rows.
- A slice like iloc[0:3] does not include the end position, so it gives positions 0, 1 and 2.
- Negative numbers count from the end, so -1 is the last row.
- Column positions are also numbers (0 for the first column).

### Step 2 – Demonstrate the Concept

#### Business Example
A manager wants to see the first 3 customers and the last customer in the table. With iloc this is done by position: iloc[0:3] for the first three and iloc[-1] for the last one, and the row labels do not matter.

### Step 3 – Implement the Concept

```python
print("First row:")
print(customers.iloc[0])

print()
print("Last row:")
print(customers.iloc[-1])

# End position is not included
print()
print(customers.iloc[0:3])

# Specific rows
print()
print(customers.iloc[[0, 2, 4]])
```

```python
# Rows and columns by position
print(customers.iloc[0:3, [1, 2]])

print()
print(customers.iloc[:, 0:3])

print()
print("Value at row 3, column 1:", customers.iloc[3, 1])

# loc and iloc with the same slice
print()
print("loc[0:3] gives", len(customers.loc[0:3]), "rows")
print("iloc[0:3] gives", len(customers.iloc[0:3]), "rows")
```

```python
# ML-style: split the rows into train and test by position
train = customers.iloc[:9]
test = customers.iloc[9:]
print("Train shape:", train.shape)
print("Test shape :", test.shape)

# Feature columns by position (Age, Income, Purchases)
X = customers.iloc[:, [2, 5, 6]]
print()
print(X.columns.tolist())
```

#### What does this program do?
In the first cell, iloc[0] gives the first row (Asha) and iloc[-1] gives the last row (Sameer). iloc[0:3] gives rows at the positions 0, 1 and 2, so it is 3 rows (Asha, Ravi and Meena), and iloc[[0, 2, 4]] gives the rows at the positions 0, 2 and 4.

In the second cell, iloc[0:3, [1, 2]] gives 3 rows and the columns at the positions 1 and 2 (Name and Age). iloc[:, 0:3] gives all rows but only the first three columns (CustomerID, Name and Age), and iloc[3, 1] gives the single value Kiran. At the end the same slice 0:3 is used with loc and iloc. loc gives 4 rows because the end is included, and iloc gives 3 rows because the end is not included.

In the last cell the table is split by position into train (the first 9 rows, shape (9, 8)) and test (the last 3 rows, shape (3, 8)). The feature columns at the positions 2, 5 and 6 are Age, Income and Purchases.

#### AI/ML Example
iloc is used for splitting data by position, for example the first 80 percent of rows for training and the rest for testing, which is common for time ordered data where the order must not be mixed. It is also used to select all feature columns except the last one (iloc[:, :-1]) and the last column as the target (iloc[:, -1]).

#### Advantages
- Works the same for any index, so the labels do not matter.
- Same slicing rules as Python lists and NumPy.
- Good for splitting data by position.

#### Limitations
- Position numbers are less readable than names, and the code breaks if the order of the columns changes.
- A position outside the table gives an IndexError.

#### Key Point
iloc selects by position (starting from 0), and the end of a slice is not included. Format: df.iloc[rows, columns].

## Topic 3 – Boolean Indexing

### Step 1 – Understand the Concept
Boolean indexing means selecting rows using True and False values. The steps are:

1. Write a condition on a column, like df["Income"] > 60. This gives a Series of True and False, one value for each row (this is called a mask).
2. Put the mask inside the brackets, df[mask], and only the rows with True are kept.

The mask can be saved in a variable, and summing a mask (mask.sum()) counts the True values. The symbol ~ reverses a mask (True becomes False and False becomes True).

If a value is missing (NaN), a comparison with it gives False, so rows with missing values are never selected by a condition like "greater than".

### Step 2 – Demonstrate the Concept

#### Business Example
A company wants to find its high income customers, which means everyone with an income above 60 thousand. The condition is written on the Income column, and only the rows that match are kept.

### Step 3 – Implement the Concept

```python
# Step 1: the mask
mask = customers["Income"] > 60
print(mask)

# Step 2: using the mask
print()
print(customers[mask])

print()
print("Number of customers with income above 60:", mask.sum())
```

```python
# The condition can also be written directly inside the brackets
older = customers[customers["Age"] > 30]
print(older[["Name", "Age"]])
print()
print("Rows with Age above 30:", len(older))

# Selecting rows and columns together with loc
print()
print(customers.loc[customers["Income"] > 60, ["Name", "Income"]])

# The ~ symbol reverses the condition
print()
not_hyderabad = customers[~(customers["City"] == "Hyderabad")]
print("Rows outside Hyderabad:", len(not_hyderabad))
```

#### What does this program do?
In the first cell, the condition customers["Income"] > 60 creates a mask of 12 True or False values. The output has True for Ravi, Kiran, Rahul, Neha and Vikram, and False for all the others. Arjun is also False, because his income is missing and a comparison with NaN is always False. Putting the mask in the brackets keeps these 5 rows, and mask.sum() prints 5.

In the second cell, the condition is written directly inside the brackets for Age above 30. It gives 7 rows (Ravi, Kiran, Sana, Rahul, Neha, Vikram and Sameer). Divya is not in the result, because her age is missing. Then loc is used with the same kind of condition to get only the Name and Income columns of the 5 high income customers. At the end ~ is used to get the customers who are not from Hyderabad, which are 8 rows out of 12, since 4 customers are from Hyderabad.

#### AI/ML Example
Boolean indexing is used to take out rows that match a rule, like selecting only the rows of one class, removing rows with impossible values (for example Age less than 0), or removing outliers (for example Income more than some limit). Because a condition with NaN is False, missing values have to be handled separately.

#### Advantages
- Very powerful and fast for selecting rows by a rule.
- No loops are needed.
- Works with any condition on any column.

#### Limitations
- Rows with missing values are silently left out, which can be missed.
- Long conditions can become hard to read.

#### Key Point
A condition on a column gives a True/False mask, and df[mask] keeps only the True rows. ~ reverses a mask.

## Topic 4 – Filtering Rows

### Step 1 – Understand the Concept
Filtering rows means keeping only the rows that match one or more conditions. To join many conditions, these operators are used:

- & : and (both conditions must be True)
- | : or (at least one condition must be True)
- ~ : not (reverses the condition)

Every single condition must be inside round brackets, like (df["Age"] > 30) & (df["Gender"] == "M"), because of the way Python reads these operators. The words and, or and not cannot be used on a column, because they give an error.

### Step 2 – Demonstrate the Concept

#### Business Example
The marketing team wants to send an offer only to Gold customers who live in Hyderabad. This needs two conditions together: the city is Hyderabad and the segment is Gold.

### Step 3 – Implement the Concept

```python
# AND: both conditions
hyd_gold = customers[(customers["City"] == "Hyderabad") & (customers["Segment"] == "Gold")]
print(hyd_gold[["Name", "City", "Segment"]])

# OR: any one of the conditions
mumbai_pune = customers[(customers["City"] == "Mumbai") | (customers["City"] == "Pune")]
print()
print(mumbai_pune[["Name", "City"]])

# NOT
males = customers[~(customers["Gender"] == "F")]
print()
print("Rows that are not F:", len(males))
```

```python
# Three conditions together
target_group = customers[(customers["Age"] > 30) &
                         (customers["Income"] >= 60) &
                         (customers["Gender"] == "M")]
print(target_group[["Name", "Age", "Income", "Gender"]])
print()
print("Rows in the target group:", len(target_group))

# Using and instead of & gives an error
try:
    customers[(customers["Age"] > 30) and (customers["Income"] > 60)]
except ValueError as error:
    print()
    print("ValueError:", error)
```

#### What does this program do?
In the first cell, the AND condition gives only Kiran and Vikram, because they are the only customers who are from Hyderabad and are also Gold. The OR condition gives the 5 customers who are from Mumbai or Pune (Meena, Sana, Rahul, Neha and Sameer). The NOT condition keeps the rows where the gender is not F, which is 6 rows.

In the second cell, three conditions are joined with &: age above 30, income of 60 or more, and gender M. The result has 4 rows (Ravi, Kiran, Rahul and Vikram). At the end, the same type of condition is written with the word and, and Python gives a ValueError saying that the truth value of a Series is ambiguous. This shows why & has to be used with brackets and not the word and.

#### AI/ML Example
Filtering rows is used to build a subset of the data for a specific group, for example training a separate model only for high value customers, or removing rows that do not follow the rules of the problem. In data cleaning, several conditions are joined to remove wrong rows, like negative ages or incomes that are too high.

#### Advantages
- Many conditions can be joined into one rule.
- Works on the whole table at once, with no loop.
- The result is a normal DataFrame that can be used for the next step.

#### Limitations
- Forgetting the brackets or using and/or gives errors or wrong results.
- Rows with missing values in the condition columns are left out.
- Long filters are hard to read (query() can help with this).

#### Key Point
Join conditions with & (and), | (or) and ~ (not), and put every condition in brackets.

## Topic 5 – query()

### Step 1 – Understand the Concept
query() filters rows using a condition written as a text string, like df.query("Income > 60"). It does the same job as boolean indexing, but the code is shorter and easier to read.

- Inside the string, column names are written directly, without df[...].
- The words and, or and not can be used, so no extra brackets are needed.
- A Python variable can be used inside the string by putting @ before its name, like "Income > @min_income".
- Text values are written inside quotes within the string, for example "City == 'Pune'".
- in can be used to check against a list.

### Step 2 – Demonstrate the Concept

#### Business Example
A manager who knows a little about data wants to write the filter almost like a sentence: "Income greater than 60". With query() this is written exactly like that, so it is easy for others in the team to read.

### Step 3 – Implement the Concept

```python
# A simple condition
print(customers.query("Income > 60")[["Name", "Income"]])

# More conditions with and
print()
print(customers.query("City == 'Hyderabad' and Segment == 'Gold'")[["Name", "City", "Segment"]])

print()
print(customers.query("Age > 30 and Income >= 60 and Gender == 'M'")[["Name", "Age", "Income"]])
```

```python
# Using Python variables with @
min_income = 60
print(customers.query("Income > @min_income")[["Name", "Income"]])

# Checking against a list with in
cities = ["Mumbai", "Pune"]
print()
print(customers.query("City in @cities")[["Name", "City"]])

# Same result as boolean indexing
a = customers.query("Income > 60")
b = customers[customers["Income"] > 60]
print()
print("query and boolean indexing give the same result?", a.equals(b))
```

#### What does this program do?
The first cell uses query() with a simple condition, which gives the 5 customers with an income above 60 (Ravi, Kiran, Rahul, Neha and Vikram). Then it uses conditions joined with and. The Hyderabad and Gold query gives Kiran and Vikram, and the query with three conditions gives Ravi, Kiran, Rahul and Vikram, the same as the results of the filtering topic.

In the second cell, the variable min_income is used inside the query with @, and it gives the same 5 customers as before. The variable cities is used with in, and it gives the 5 customers from Mumbai or Pune. At the end, the result of query() is compared with the result of boolean indexing using equals(), and the output is True, which shows that both ways give the same table.

#### AI/ML Example
In Exploratory Data Analysis, many quick filters are tried one after another, and query() makes them short and readable. The @ variable lets the same query be used with different values, such as different limits or lists of categories, for example when checking the data for many groups in a loop.

#### Advantages
- Short, clean and easy to read, especially for many conditions.
- No repeated df[...] and no extra brackets.
- Can use Python variables with @.

#### Limitations
- The condition is a string, so mistakes (like a wrong column name) are only found when the code runs, and editors cannot check it.
- Column names with spaces or special characters are harder to write.
- For very small tables it is not faster than boolean indexing.

#### Key Point
query("condition") filters rows using a text condition. Use @variable for Python variables and quotes for text values.

## Topic 6 – isin()

### Step 1 – Understand the Concept
isin() checks if every value of a column is present in a given list, and gives True or False for each row. The result is a mask that can be used to filter rows. It is a short way of writing many "equal to" conditions joined with or.

~ before the condition reverses it, which gives the rows whose values are not in the list.

### Step 2 – Demonstrate the Concept

#### Business Example
A company wants to look at the customers from only two cities, Mumbai and Pune, for a regional campaign. Instead of writing the condition for Mumbai or Pune, it writes isin(["Mumbai", "Pune"]), and if more cities are needed later, only the list changes.

### Step 3 – Implement the Concept

```python
# Customers from the listed cities
selected = customers[customers["City"].isin(["Mumbai", "Pune"])]
print(selected[["Name", "City"]])

# Customers whose segment is NOT Bronze
not_bronze = customers[~customers["Segment"].isin(["Bronze"])]
print()
print("Rows that are not Bronze:", len(not_bronze))

# Selecting by a list of IDs
ids = [2, 4, 8]
print()
print(customers[customers["CustomerID"].isin(ids)][["CustomerID", "Name"]])
```

```python
# isin gives True/False, so it can be counted
print("Gold or Silver customers:", customers["Segment"].isin(["Gold", "Silver"]).sum())

# isin is the same as many == conditions joined with |
a = customers["City"].isin(["Mumbai", "Pune"])
b = (customers["City"] == "Mumbai") | (customers["City"] == "Pune")
print("Same result as the long condition?", a.equals(b))
```

#### What does this program do?
In the first cell, isin(["Mumbai", "Pune"]) keeps the 5 customers from these two cities (Meena, Sana, Rahul, Neha and Sameer). With ~ in front, the second filter removes the Bronze customers. There are 3 Bronze customers, so 9 rows remain. The third filter selects customers by a list of IDs and gives Ravi, Kiran and Rahul (IDs 2, 4 and 8).

In the second cell, summing the isin() result counts the Gold and Silver customers, which is 9 (4 Gold and 5 Silver). The last check shows that isin() gives exactly the same mask as the longer condition with two == and one |, so the output is True.

#### AI/ML Example
isin() is used to keep only some classes or categories for a model, for example only Gold and Silver, and to remove unwanted or invalid categories with ~. It is also used to select the rows of a given list of IDs, such as the IDs picked for a test set, or to check which items of one dataset are present in another.

#### Advantages
- Shorter and clearer than many == conditions joined by |.
- The list can be changed or created by the program.
- ~ gives the "not in the list" filter easily.

#### Limitations
- Checks only exact matches, so it cannot find partial text (string methods are needed for that).
- It is case-sensitive for text, so "pune" does not match "Pune".

#### Key Point
isin(list) gives True for the values that are in the list. Use ~ for the values that are not in the list.

## Topic 7 – between()

### Step 1 – Understand the Concept
between(left, right) checks if the values of a column lie in a range, and gives True or False for each row. By default both ends are included, so it is the same as (x >= left) & (x <= right).

The inclusive option can change this:
- "both" : both ends included (default)
- "neither" : both ends excluded
- "left" : only the left end included
- "right" : only the right end included

Missing values give False.

### Step 2 – Demonstrate the Concept

#### Numerical Example
Take the ages of the customers. The ages between 30 and 40, with both 30 and 40 included, are 32, 36, 31, 38 and 34, which is 5 customers. An age of 41 or 29 is outside the range.

### Step 3 – Implement the Concept

```python
# Age between 30 and 40 (both included)
group = customers[customers["Age"].between(30, 40)]
print(group[["Name", "Age"]])

# The inclusive option
print()
print("between(31, 38) both ends included :", customers["Age"].between(31, 38).sum())
print("between(31, 38) neither included   :", customers["Age"].between(31, 38, inclusive="neither").sum())
```

```python
# Income between 50 and 70
middle = customers[customers["Income"].between(50, 70)]
print(middle[["Name", "Income"]])

# between is the same as two comparisons joined with &
a = customers["Income"].between(50, 70)
b = (customers["Income"] >= 50) & (customers["Income"] <= 70)
print()
print("Same as the long condition?", a.equals(b))
```

#### What does this program do?
In the first cell, between(30, 40) keeps the customers with an age from 30 to 40, which are 5 rows (Ravi 32, Sana 36, Neha 31, Vikram 38 and Sameer 34). Divya is not included because her age is missing. After that, between(31, 38) is used with the two inclusive settings, and the number of True values is counted. With both ends included, the count is 5 (31, 32, 34, 36 and 38). With inclusive="neither", the ages 31 and 38 are dropped, so the count is 3 (32, 34 and 36).

In the second cell, between(50, 70) selects the customers with an income from 50 to 70, which are 6 rows (Ravi, Meena, Divya, Neha, Vikram and Sameer). The last check shows that between() gives the same result as (Income >= 50) & (Income <= 70), so the output is True.

#### AI/ML Example
between() is used to keep only the values in a valid range, for example removing wrong ages by keeping only the ages between 0 and 100, or removing outliers outside a chosen limit. It is also used to create age groups or income bands, and to select a date range in time series data.

#### Advantages
- Cleaner than writing two comparisons with &.
- The inclusive option gives control of the ends.
- Works with numbers and also with dates.

#### Limitations
- Only for one column and one continuous range. Many separate ranges need more conditions.
- Rows with missing values are left out.

#### Key Point
between(left, right) checks a range, and both ends are included by default. Use inclusive to change that.

## Topic 8 – Sorting Data

### Step 1 – Understand the Concept
Sorting arranges the rows in an order.

- sort_values("column") sorts by the values of a column, smallest first by default. ascending=False gives the largest first.
- A list of columns sorts by many columns. The first column is the main one, and the next ones are used when there is a tie. ascending can also be a list, like [True, False].
- Missing values are placed at the end by default. na_position="first" puts them at the top.
- ignore_index=True gives new row numbers after sorting.
- sort_index() sorts by the row labels, and it can bring back the original order.
- nlargest(n, "column") and nsmallest(n, "column") give the top or bottom n rows directly.

sort_values() returns a new sorted table, and the original table does not change.

### Step 2 – Demonstrate the Concept

#### Business Example
A company wants a ranking of its customers by income, starting with the highest. Sorting by Income in descending order gives this list, and the top few rows are the most valuable customers.

### Step 3 – Implement the Concept

```python
# Smallest income first (missing value goes to the end)
print(customers.sort_values("Income")[["Name", "Income"]])

# Highest income first
print()
by_income = customers.sort_values("Income", ascending=False)
print(by_income[["Name", "Income"]])

# Top 3 customers by income
print()
print(by_income[["Name", "Income"]].head(3))
```

```python
# Sorting by two columns: City A to Z, and inside each city the highest income first
print(customers.sort_values(["City", "Income"], ascending=[True, False])[["City", "Name", "Income"]])

# Missing values first
print()
print(customers.sort_values("Income", na_position="first")[["Name", "Income"]].head(3))

# Fresh row numbers after sorting
print()
print(customers.sort_values("Age", ignore_index=True)[["Name", "Age"]].head(3))
```

```python
# Sorting text, and going back to the original order with sort_index
by_name = customers.sort_values("Name")
print(by_name[["Name"]].head(3))

print()
print("Back to the original order?", by_name.sort_index().equals(customers))

# nlargest and nsmallest
print()
print(customers.nlargest(3, "Purchases")[["Name", "Purchases"]])
print()
print(customers.nsmallest(3, "Purchases")[["Name", "Purchases"]])
```

#### What does this program do?
In the first cell, sort_values("Income") puts the customers in the order of the lowest income first (Pooja 38, Asha 40, Sana 45 and so on up to Rahul 80), and Arjun, whose income is missing, is at the end. With ascending=False the order starts with Rahul 80, then Kiran 72, Vikram 70, Ravi 65 and so on, and Arjun is still at the end. The head(3) of this sorted table gives the top 3 customers by income, which are Rahul, Kiran and Vikram.

In the second cell, the sort by City (A to Z) and then by Income (highest first) groups the customers by city (Chennai, Hyderabad, Mumbai and Pune), and inside each city the highest income is on top. For example, in Pune the order is Rahul 80, Meena 58 and Sameer 55. With na_position="first" Arjun comes to the top. With ignore_index=True the row numbers are 0, 1, 2 again, and the youngest customers come first (Asha 25, Pooja 27 and Arjun 28).

In the last cell, sorting by Name puts the names in the alphabetical order, starting with Arjun, Asha and Divya. Using sort_index() on the sorted table brings back the original order, so equals() prints True. nlargest(3, "Purchases") gives Rahul 15, Kiran 12 and Vikram 10, and nsmallest(3, "Purchases") gives Pooja 1, Arjun 2 and Asha 3.

#### AI/ML Example
Sorting is used to rank things, like the customers with the highest value, or the features with the highest scores, and to look at the largest and smallest values to find outliers. In time series data the rows must be sorted by date before any calculation. nlargest() is a quick way to find the top n results, for example the best models in a list of results.

#### Advantages
- Easy ranking and ordering of data, also by many columns.
- Handles missing values with na_position.
- The original table does not change, so the data stays safe.

#### Limitations
- Sorting a very large table can be slow and uses more memory.
- The row labels stay with their rows, so the index looks mixed up unless ignore_index=True or reset_index() is used.
- The result must be saved in a variable, otherwise the sorted table is lost.

#### Key Point
sort_values("column") sorts by the values. Use ascending=False for the largest first, and a list of columns to sort by many columns.

## Interview Questions

#### 1. What is the difference between loc and iloc?

loc selects by labels (names) and includes the end of a slice. iloc selects by position numbers starting from 0 and does not include the end of a slice. For example, with the default index, df.loc[0:3] gives 4 rows but df.iloc[0:3] gives 3 rows.

#### 2. How do you filter rows with more than one condition? Why can't we use and / or?

Join the conditions with & (and), | (or) and ~ (not), and put each condition in brackets, for example df[(df["Age"] > 30) & (df["Gender"] == "M")]. The words and / or work on single True/False values, but a column is a whole Series of values, so pandas gives a ValueError.

#### 3. What is the difference between boolean indexing and query()?

Both filter rows and give the same result. Boolean indexing writes the condition with the column names, like df[df["Income"] > 60]. query() takes the condition as a string, like df.query("Income > 60"), which is shorter and easier to read, and it can use variables with @. The disadvantage of query() is that the condition is a string, so mistakes are found only when it runs.

#### 4. What do isin() and between() do?

isin(list) gives True for the values that are in the list, and it is a short form of many == conditions joined with |. between(a, b) gives True for the values in the range from a to b, with both ends included by default, and it is the same as (x >= a) & (x <= b).

#### 5. How do you sort a DataFrame by more than one column, and what happens to missing values?

Use df.sort_values(["col1", "col2"], ascending=[True, False]). The first column is the main sort and the second one is used for ties. Missing values are put at the end by default, and na_position="first" puts them at the top. The original table does not change, so the result must be saved in a variable.
