### Sample Data
Instead of creating new data for every topic, I will reuse a few small arrays. `sales` is the daily sales of one shop for a week, and `store_sales` is the sales of 3 stores over 4 days (rows = stores, columns = days).

## Topic 1 – Sum

### Step 1 – Understand the Concept
`np.sum()` adds up all the elements of an array. For a 2D array, `axis` can be used to add down the columns (`axis=0`) or across the rows (`axis=1`).

### Step 2 – Demonstrate the Concept

#### Business Example
A shop wants its total sales for the week. With the daily sales 120, 135, 150, 110, 180, 210 and 190, the total is 1095 units. For the 3-store table, the shop owner may want the total of each store (across the days) and the total of each day (across the stores).

### Step 3 – Implement the Concept

```python
print("Total weekly sales :", np.sum(sales))
print("Same using method  :", sales.sum())

print("\nTotal of all stores and days:", store_sales.sum())
print("Total per day   (axis=0)    :", store_sales.sum(axis=0))
print("Total per store (axis=1)    :", store_sales.sum(axis=1))

# ML-style: sum of squared errors
y_true = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.5, 5.0, 4.0, 8.0])
sse = np.sum((y_pred - y_true) ** 2)
print("\nSum of squared errors:", sse)
```

#### What does this program do?
The program adds all the daily sales, then adds the 2D sales table in three ways: all values, per day and per store. At the end it calculates the sum of squared errors of some predictions.

`sum()` is used because it adds a whole array in one step. The output shows `1095` for the week, `[410 415 455 405]` per day, `[515 780 390]` per store, and `3.5` for the sum of squared errors.

#### AI/ML Example
Loss functions such as Sum of Squared Errors (SSE) and Mean Squared Error (MSE) are built using `np.sum`. The model's errors for all samples are squared and added to get one number that tells how good the model is.

#### Advantages
- One call adds any number of values, in any dimension.
- `axis` gives row-wise or column-wise totals easily.

#### Limitations
- If the array contains `nan` (missing values), the result is `nan`. `np.nansum()` should be used in that case.

#### Key Point
`np.sum()` adds the elements. `axis=0` gives column totals and `axis=1` gives row totals.

## Topic 2 – Mean

### Step 1 – Understand the Concept
The mean is the average of the values: the sum of all values divided by the number of values. `np.mean()` calculates it and also supports `axis`.

### Step 2 – Demonstrate the Concept

#### Numerical Example
The marks of five students are 60, 70, 75, 80 and 90.

Mean = (60 + 70 + 75 + 80 + 90) / 5 = 375 / 5 = 75

### Step 3 – Implement the Concept

```python
marks = np.array([60, 70, 75, 80, 90])
print("Mean marks:", np.mean(marks))

print("\nAverage sales per store (axis=1):", store_sales.mean(axis=1))
print("Average sales per day   (axis=0):", np.round(store_sales.mean(axis=0), 2))

# ML-style: mean of each feature (column) in a small dataset
X = np.array([[22, 25000],
              [25, 30000],
              [30, 45000],
              [35, 50000],
              [40, 60000]])      # age, salary
print("\nFeature means [age, salary]:", X.mean(axis=0).tolist())
```

#### What does this program do?
The program calculates the mean of the marks, then the mean of each store and each day from the sales table, and finally the mean of each feature (age and salary) in a small dataset.

`mean()` is used to find the typical value of the data. The output shows `75.0` for the marks, `[128.75 195. 97.5]` per store, `[136.67 138.33 151.67 135.]` per day and `[30.4, 42000.0]` for the features.

#### AI/ML Example
The mean of every feature is used for centering and scaling the data before training a model (for example `X - X.mean(axis=0)`). The mean is also used in loss functions like Mean Squared Error.

#### Advantages
- Simple and easy to understand.
- Uses every value in the data.

#### Limitations
- Extreme values (outliers) can pull the mean away from the typical value. The median handles this better.
- The result is `nan` if there are missing values (use `np.nanmean()`).

#### Key Point
Mean = sum of values ÷ number of values. It is affected by outliers.

## Topic 3 – Median

### Step 1 – Understand the Concept
The median is the middle value of the data when it is sorted. If there is an even number of values, it is the average of the two middle values. `np.median()` calculates it.

The median is not affected much by extreme values, so it is a better "typical value" than the mean when the data has outliers.

### Step 2 – Demonstrate the Concept

#### Business Example
A small company has 5 employees with salaries 30000, 32000, 35000, 38000 and 400000 (the owner). The mean is pulled up by the owner's salary and is 107000, which does not describe a normal employee. The median is 35000, which is a much better picture of the typical salary.

### Step 3 – Implement the Concept

```python
salaries = np.array([30000, 32000, 35000, 38000, 400000])
print("Mean salary  :", np.mean(salaries))
print("Median salary:", np.median(salaries))

# Even number of values -> average of the two middle values
print("\nMedian of [1, 2, 3, 4]:", np.median([1, 2, 3, 4]))

# ML-style: filling a missing value with the median
income = np.array([30, 32, np.nan, 38, 400])
median_income = np.nanmedian(income)
filled = np.where(np.isnan(income), median_income, income)
print("\nIncome with a missing value:", income)
print("Median ignoring the missing value:", median_income)
print("After filling the missing value  :", filled)
```

#### What does this program do?
The program compares the mean and median of salaries that contain one very large value, finds the median of an even-sized array, and then fills a missing value (`nan`) in an income array using the median.

`np.nanmedian()` is used because it ignores `nan` while calculating, and `np.where()` is used to replace only the missing positions. The output shows a mean of `107000.0`, a median of `35000.0`, a median of `2.5` for `[1, 2, 3, 4]`, and the missing value replaced by `35.0`.

#### AI/ML Example
When a feature has outliers or is skewed (such as income or house price), missing values are often filled with the median instead of the mean, because the median is not pulled by extreme values.

#### Advantages
- Not affected by outliers.
- A good measure of the typical value for skewed data.

#### Limitations
- Does not use every value in the data.
- Sorting is needed internally, so it is slightly slower than the mean on very large data.

#### Key Point
Median = the middle value of sorted data. Use it instead of the mean when there are outliers.

## Topic 4 – Variance

### Step 1 – Understand the Concept
Variance tells how spread out the values are around the mean. It is calculated as the average of the squared differences from the mean.

`np.var()` calculates it. By default it gives the **population variance** (divides by n). For a **sample variance** (divides by n − 1), use `ddof=1`.

### Step 2 – Demonstrate the Concept

#### Numerical Example
Take the values 2, 4, 4, 4, 5, 5, 7, 9. The mean is 5.

Differences from the mean: -3, -1, -1, -1, 0, 0, 2, 4

Squared differences: 9, 1, 1, 1, 0, 0, 4, 16 → sum = 32

Population variance = 32 / 8 = 4. Sample variance = 32 / 7 = 4.571

### Step 3 – Implement the Concept

```python
data = np.array([2, 4, 4, 4, 5, 5, 7, 9])

print("Mean                :", data.mean())
print("Population variance :", np.var(data))
print("Sample variance     :", np.var(data, ddof=1))

# Checking the formula manually
manual = np.mean((data - data.mean()) ** 2)
print("Manual calculation  :", manual)

# ML-style: variance of each feature
features = np.array([[5, 1],
                     [5, 4],
                     [5, 2],
                     [5, 8],
                     [5, 5]])
print("\nVariance of each feature:", np.var(features, axis=0))
```

#### What does this program do?
The program calculates the population and sample variance of the data, and then checks the result with the formula written by hand. At the end it calculates the variance of each column in a small feature table.

`ddof=1` is used for sample variance because a sample usually underestimates the spread of the full population. The output shows `4.0` for the population variance, `4.571...` for the sample variance, `4.0` from the manual calculation, and `[0. 6.]` for the two features.

#### AI/ML Example
A feature with zero or very low variance has the same value for almost every sample (like the first feature above, which is always 5), so it gives no useful information to a model. Variance is used to remove such features. Variance is also the base of PCA and many other methods.

#### Advantages
- Gives one number for how spread out the data is.
- Used inside many statistical and ML methods.

#### Limitations
- The unit is squared (for example "rupees squared"), so it is hard to interpret directly. Standard deviation fixes this.
- Strongly affected by outliers.

#### Key Point
Variance = average of squared differences from the mean. Use `ddof=1` for the sample variance.

## Topic 5 – Standard Deviation

### Step 1 – Understand the Concept
Standard deviation is the square root of the variance. It tells how far the values are from the mean on average, and it is in the same unit as the data, so it is easier to understand than variance.

`np.std()` calculates it. As with variance, `ddof=1` gives the sample standard deviation.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Two delivery companies both take 31 minutes on average.

- Company A delivery times: 30, 32, 31, 29, 33 (always close to 31).
- Company B delivery times: 20, 45, 25, 40, 25 (sometimes very fast, sometimes very late).

Both have the same mean, but Company B is less reliable. The standard deviation shows this difference.

### Step 3 – Implement the Concept

```python
company_a = np.array([30, 32, 31, 29, 33])
company_b = np.array([20, 45, 25, 40, 25])

print("Mean A:", company_a.mean(), "| Std A:", round(company_a.std(), 2))
print("Mean B:", company_b.mean(), "| Std B:", round(company_b.std(), 2))

# ML-style: standardisation (z-score) of the age and salary features
X = np.array([[22, 25000],
              [25, 30000],
              [30, 45000],
              [35, 50000],
              [40, 60000]])
z = (X - X.mean(axis=0)) / X.std(axis=0)
print("\nStandardised data:\n", np.round(z, 2))
print("New means:", np.round(z.mean(axis=0), 2))
print("New stds :", np.round(z.std(axis=0), 2))
```

#### What does this program do?
The program compares the standard deviation of two delivery companies with the same mean. It then standardises a small dataset (age and salary) by subtracting each column's mean and dividing by its standard deviation.

Standard deviation is used because it shows the spread in the original unit. The output shows both means are `31.0`, but the standard deviation is `1.41` for A and `9.7` for B. After standardising, each column has a mean of `0` and a standard deviation of `1`.

#### AI/ML Example
Standardisation (z-score scaling) is used before training many models, such as linear regression, logistic regression, SVM and neural networks, so that features with large values (like salary) do not dominate features with small values (like age).

#### Advantages
- Same unit as the data, so it is easy to interpret.
- Used for scaling features and detecting unusual values.

#### Limitations
- Affected by outliers.
- Gives a wrong picture for data that is not roughly symmetric.

#### Key Point
Standard deviation = square root of variance. A bigger value means the data is more spread out.

## Topic 6 – Minimum

### Step 1 – Understand the Concept
`np.min()` returns the smallest value in an array. With `axis`, it returns the smallest value of each column or each row.

### Step 2 – Demonstrate the Concept

#### Real-World Example
A weather app stores the temperatures of a week: 24, 22, 19, 21, 25, 27 and 23 °C. The minimum is 19 °C, which is the coldest day of the week.

### Step 3 – Implement the Concept

```python
temps = np.array([24, 22, 19, 21, 25, 27, 23])
print("Coldest temperature:", np.min(temps))

print("\nLowest sales of each store (axis=1):", store_sales.min(axis=1))
print("Lowest sales of each day   (axis=0):", store_sales.min(axis=0))

# ML-style: checking data for impossible values
ages = np.array([25, 31, -4, 40, 29])
print("\nMinimum age in the data:", ages.min())
if ages.min() < 0:
    print("Warning: negative age found, the data needs cleaning.")
```

#### What does this program do?
The program finds the lowest temperature, the lowest sales for each store and each day, and then checks an age column for impossible values by looking at its minimum.

`min()` is used for a quick look at the lower end of the data. The output shows `19`, `[110 180 90]` per store, `[ 90 100  95 105]` per day, and `-4` for the age with a warning message.

#### AI/ML Example
During data cleaning, the minimum is checked to find wrong values (such as a negative age or negative price). It is also used for Min-Max scaling, which is shown in the next topic.

#### Advantages
- Quick check of the lowest value.
- Helps find errors in data.

#### Limitations
- Only gives the value, not its position (use `argmin` for the position).
- Returns `nan` if the data has `nan` (use `np.nanmin()`).

#### Key Point
`np.min()` gives the smallest value. Use `axis` for row-wise or column-wise results.

## Topic 7 – Maximum

### Step 1 – Understand the Concept
`np.max()` returns the largest value in an array. With `axis`, it gives the largest value of each column or row.

### Step 2 – Demonstrate the Concept

#### Business Example
A shop wants to know its best day. From the weekly sales 120, 135, 150, 110, 180, 210 and 190, the maximum is 210 units.

### Step 3 – Implement the Concept

```python
print("Highest daily sales:", np.max(sales))

print("\nHighest sales of each store (axis=1):", store_sales.max(axis=1))
print("Highest sales of each day   (axis=0):", store_sales.max(axis=0))

# ML-style: Min-Max scaling brings values to the range 0 to 1
scaled = (sales - sales.min()) / (sales.max() - sales.min())
print("\nOriginal sales:", sales)
print("Min-Max scaled:", np.round(scaled, 2))
```

#### What does this program do?
The program finds the highest daily sales, the highest sales for each store and each day, and then scales the sales values to the range 0 to 1 using the minimum and maximum.

The formula `(x - min) / (max - min)` is used so that the smallest value becomes 0 and the largest becomes 1. The output shows `210`, `[150 210 105]` per store, `[200 180 210 190]` per day, and scaled values `[0.1 0.25 0.4 0. 0.7 1. 0.8]`.

#### AI/ML Example
Min-Max scaling is very common in ML preprocessing, especially for neural networks and distance-based algorithms like KNN. It puts all features on the same 0–1 scale.

#### Advantages
- Quick check of the highest value.
- Together with the minimum, it gives the range of the data.

#### Limitations
- One extreme value can squash all other values after Min-Max scaling.
- Only gives the value, not its position (use `argmax` for that).

#### Key Point
`np.max()` gives the largest value. Min-Max scaling uses both the minimum and the maximum.

## Topic 8 – Argmin

### Step 1 – Understand the Concept
`np.argmin()` returns the **index (position)** of the smallest value, not the value itself. With `axis`, it returns the position for each column or row. If the smallest value appears more than once, the first position is returned.

### Step 2 – Demonstrate the Concept

#### Business Example
The minimum of the weekly sales is 110, but the owner also wants to know which day it was. `argmin` gives the position 3, and day number 3 (counting from 0) is Thursday.

### Step 3 – Implement the Concept

```python
position = np.argmin(sales)
print("Position of the lowest sales:", position)
print("Lowest sales day            :", days[position], "with", sales[position], "units")

print("\nDay with lowest sales for each store (axis=1):", store_sales.argmin(axis=1))

# ML-style: finding the best epoch (lowest validation loss)
val_loss = np.array([0.90, 0.60, 0.45, 0.38, 0.41, 0.47])
best_epoch = np.argmin(val_loss) + 1      # +1 because epochs are counted from 1
print("\nValidation loss per epoch:", val_loss)
print("Best epoch:", best_epoch, "with loss", val_loss[best_epoch - 1])
```

#### What does this program do?
The program finds the position of the lowest sales and uses it to get the day name. It then finds the position of the lowest value in each row of the store table. At the end it finds the epoch with the lowest validation loss of a model.

`argmin` is used when the position matters more than the value. The output shows position `3`, `Thu` with `110` units, `[3 1 0]` for the stores, and best epoch `4` with loss `0.38`.

#### AI/ML Example
While training a model, the validation loss is recorded after every epoch. `np.argmin` finds the epoch with the lowest loss, which is the best point to stop training or to pick the saved model. It is also used to find the closest point in KNN (the smallest distance).

#### Advantages
- Gives the position, which can be used to look up names, labels or other arrays.

#### Limitations
- If the minimum appears several times, only the first position is returned.

#### Key Point
`argmin` gives the index of the smallest value, while `min` gives the value itself.

## Topic 9 – Argmax

### Step 1 – Understand the Concept
`np.argmax()` returns the **index (position)** of the largest value. With `axis`, it returns the position for each row or column. If the largest value appears more than once, the first position is returned.

### Step 2 – Demonstrate the Concept

#### Business Example
A store has the revenue of four products: Pen 500, Book 360, Bag 560 and Bottle 400. `argmax` gives position 2, which is the Bag, the best-selling product.

### Step 3 – Implement the Concept

```python
products = np.array(["Pen", "Book", "Bag", "Bottle"])
revenue_each = np.array([500, 360, 560, 400])

best = np.argmax(revenue_each)
print("Position of the highest revenue:", best)
print("Best product:", products[best])

# ML-style: predicted class from probabilities
classes = np.array(["cat", "dog", "bird"])
probs = np.array([[0.10, 0.70, 0.20],
                  [0.50, 0.30, 0.20],
                  [0.20, 0.20, 0.60]])     # 3 samples, probability of each class

predicted_index = np.argmax(probs, axis=1)
print("\nPredicted class positions:", predicted_index)
print("Predicted classes        :", classes[predicted_index])
```

#### What does this program do?
The program finds the best-selling product using `argmax` and the product names array. It then takes the class probabilities for 3 samples and uses `argmax` along `axis=1` to pick the class with the highest probability for each sample.

`argmax` is used when we need the position of the largest value, not the value. The output shows position `2`, best product `Bag`, positions `[1 0 2]` and predicted classes `['dog' 'cat' 'bird']`.

#### AI/ML Example
This is how a classification model gives its final answer. The model outputs a probability for each class, and `np.argmax` picks the class with the highest probability as the prediction. It is used in image classifiers, text classifiers and many other models.

#### Advantages
- Converts model probabilities into a final class in one line.
- Works on a whole batch of samples at once with `axis=1`.

#### Limitations
- If two values are equal and highest, only the first position is returned.

#### Key Point
`argmax` gives the index of the largest value. In classification, it gives the predicted class.

## Topic 10 – Percentile

### Step 1 – Understand the Concept
A percentile tells the value below which a given percentage of the data falls. For example, the 90th percentile is the value below which 90% of the data lies.

`np.percentile(array, q)` calculates it, where `q` is between 0 and 100. The 25th, 50th and 75th percentiles are called Q1, Q2 (the median) and Q3. The difference Q3 − Q1 is called the Interquartile Range (IQR). When a percentile is not exactly at a data point, NumPy gives a value between the two nearest points.

### Step 2 – Demonstrate the Concept

#### Real-World Example
In an exam, a student's mark is at the 90th percentile. This means the student scored higher than 90% of the students in the class.

### Step 3 – Implement the Concept

```python
exam_marks = np.array([35, 42, 48, 55, 60, 62, 68, 71, 75, 80, 85, 92])

print("25th percentile (Q1):", np.percentile(exam_marks, 25))
print("50th percentile     :", np.percentile(exam_marks, 50))
print("75th percentile (Q3):", np.percentile(exam_marks, 75))
print("90th percentile     :", np.percentile(exam_marks, 90))
print("Several at once     :", np.percentile(exam_marks, [25, 50, 75]))

# ML-style: finding outliers using the IQR rule
ages = np.array([22, 25, 24, 23, 26, 27, 95, 24, 25, 26])
q1, q3 = np.percentile(ages, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outliers = ages[(ages < lower) | (ages > upper)]

print("\nQ1:", q1, "| Q3:", q3, "| IQR:", iqr)
print("Allowed range:", lower, "to", upper)
print("Outliers     :", outliers)
```

#### What does this program do?
The program finds the 25th, 50th, 75th and 90th percentiles of exam marks, and shows that several percentiles can be calculated in one call. It then uses Q1, Q3 and the IQR to find outliers in an age column.

The IQR rule (anything below Q1 − 1.5×IQR or above Q3 + 1.5×IQR is an outlier) is used because it is not affected by the extreme values it is looking for. The output shows `53.25`, `65.0`, `76.25` and `84.5` for the marks. For the ages, Q1 is `24.0`, Q3 is `26.0`, the allowed range is `21.0` to `29.0`, and the outlier is `[95]`.

#### AI/ML Example
Percentiles are used in outlier detection, in capping extreme values at a chosen percentile (for example the 1st and 99th), and in checking how a model's error is spread (such as the 95th percentile of the prediction error).

#### Advantages
- Not affected much by extreme values.
- Describes the whole distribution, not just the centre.

#### Limitations
- Needs enough data to be meaningful.
- Different methods of calculating percentiles can give slightly different values.

#### Key Point
Percentile = the value below which a given % of data lies. The 50th percentile is the median, and IQR = Q3 − Q1.

## Topic 11 – Dot Product

### Step 1 – Understand the Concept
The dot product of two 1D arrays of the same length is found by multiplying the matching elements and adding all the results.

For `a = [a1, a2, a3]` and `b = [b1, b2, b3]`: dot product = a1×b1 + a2×b2 + a3×b3

`np.dot(a, b)` or `a @ b` can be used. The result is a single number.

### Step 2 – Demonstrate the Concept

#### Numerical Example
For `a = [1, 2, 3]` and `b = [4, 5, 6]`:

Dot product = (1×4) + (2×5) + (3×6) = 4 + 10 + 18 = 32

### Step 3 – Implement the Concept

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("np.dot(a, b)      :", np.dot(a, b))
print("a @ b             :", a @ b)
print("np.sum(a * b)     :", np.sum(a * b))      # same result, written step by step

# ML-style: one neuron = weighted sum of inputs + bias
inputs = np.array([0.5, 0.8, 0.2])
weights = np.array([0.4, 0.3, 0.9])
bias = 0.1
neuron_output = np.dot(inputs, weights) + bias
print("\nNeuron output:", round(neuron_output, 2))
```

#### What does this program do?
The program calculates the dot product of two arrays in three ways, which all give the same result. It then calculates the output of a single neuron by taking the dot product of inputs and weights and adding a bias.

The dot product is used because it combines many multiplications and additions in one operation. The output shows `32` three times, and `0.72` for the neuron (0.5×0.4 + 0.8×0.3 + 0.2×0.9 + 0.1).

#### AI/ML Example
Every neuron in a neural network calculates a dot product of its inputs and weights (plus a bias). The dot product is also used in linear regression predictions and in finding how similar two vectors are (cosine similarity).

#### Advantages
- One fast operation instead of a loop.
- The base of nearly all ML model calculations.

#### Limitations
- Both arrays must have the same length, otherwise NumPy gives a `ValueError`.

#### Key Point
Dot product = multiply matching elements and add them. It is the core calculation of a neuron.

## Topic 12 – Matrix Multiplication

### Step 1 – Understand the Concept
Matrix multiplication combines two 2D arrays. Each element of the result is the dot product of a row from the first matrix and a column from the second matrix.

The rule is that the **number of columns of the first matrix must be equal to the number of rows of the second matrix**. A matrix of shape `(m, n)` multiplied by a matrix of shape `(n, p)` gives a result of shape `(m, p)`.

`np.matmul(A, B)` or `A @ B` is used. The `*` operator does **element-wise** multiplication, which is different.

### Step 2 – Demonstrate the Concept

#### Business Example
Two stores sell three products. The table of units sold is a 2×3 matrix and the prices form a 3×1 vector. Multiplying them gives the total revenue of each store.

- Store 1: 10×50 + 5×120 + 8×80 = 1740
- Store 2: 6×50 + 9×120 + 4×80 = 1700

### Step 3 – Implement the Concept

```python
units = np.array([[10, 5, 8],
                  [ 6, 9, 4]])        # 2 stores x 3 products
prices = np.array([50, 120, 80])      # price of each product

print("Revenue of each store:", units @ prices)

# Element-wise (*) vs matrix multiplication (@)
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("\nA * B (element-wise):\n", A * B)
print("A @ B (matrix multiplication):\n", A @ B)

# ML-style: a layer of a neural network, outputs = X @ W
X = np.array([[1, 2],
              [3, 4],
              [5, 6]])                # 3 samples, 2 features
W = np.array([[0.5,  1.0],
              [0.5, -1.0]])           # 2 features -> 2 outputs
print("\nX shape:", X.shape, "| W shape:", W.shape)
print("X @ W:\n", X @ W)
print("Result shape:", (X @ W).shape)

# Shapes that do not match
try:
    np.ones((2, 3)) @ np.ones((2, 3))
except ValueError as error:
    print("\nError:", error)
```

#### What does this program do?
The program multiplies a units table by a price vector to get each store's revenue. It then shows the difference between `*` and `@` on two small matrices, applies a layer of a neural network (`X @ W`) to 3 samples, and finally shows the error when the shapes do not match.

`@` is used because it follows the matrix multiplication rule, while `*` only multiplies matching positions. The output shows `[1740 1700]`, an element-wise result of `[[5 12] [21 32]]`, a matrix result of `[[19 22] [43 50]]`, an `X @ W` result with shape `(3, 2)`, and a `ValueError` for shapes `(2,3)` and `(2,3)`.

#### AI/ML Example
A layer of a neural network calculates `outputs = X @ W + b` for all samples at once. Matrix multiplication is the most common operation in deep learning, and GPUs are built to do it fast.

#### Advantages
- Processes many samples and many neurons in one operation.
- Very fast and highly optimised.

#### Limitations
- The shapes must follow the rule (columns of the first = rows of the second).
- The order matters: `A @ B` is usually not equal to `B @ A`.

#### Key Point
Matrix multiplication: `(m, n) @ (n, p)` gives `(m, p)`. Use `@` for matrix multiplication and `*` for element-wise multiplication.

## Topic 13 – Transpose

### Step 1 – Understand the Concept
Transpose flips an array over its diagonal: rows become columns and columns become rows. A matrix of shape `(m, n)` becomes `(n, m)`.

It can be done with `array.T` or `np.transpose(array)`. A 1D array is not changed by transpose. The transpose is a **view**, not a copy.

### Step 2 – Demonstrate the Concept

#### Numerical Example
For the matrix `[[1, 2, 3], [4, 5, 6]]` with shape (2, 3), the transpose is `[[1, 4], [2, 5], [3, 6]]` with shape (3, 2).

### Step 3 – Implement the Concept

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Original shape:", M.shape)
print("Transpose M.T :\n", M.T)
print("Transposed shape:", M.T.shape)
print("np.transpose gives the same:", np.array_equal(M.T, np.transpose(M)))
print("Transpose shares memory with the original:", np.shares_memory(M, M.T))

# ML-style: X.T @ X, used in linear regression
X = np.array([[1, 2],
              [3, 4],
              [5, 6]])                # 3 samples, 2 features
print("\nX.T @ X:\n", X.T @ X)
print("Shape of X.T @ X:", (X.T @ X).shape)
```

#### What does this program do?
The program transposes a 2×3 matrix and prints the new shape. It checks that `.T` and `np.transpose` give the same result and that the transpose shares memory with the original. It then calculates `X.T @ X` for a small dataset.

Transpose is used to make shapes fit for matrix multiplication. The output shows a `(3, 2)` array for the transpose, `True` for both checks, and `X.T @ X` as `[[35 44] [44 56]]` with shape `(2, 2)`, which is features × features.

#### AI/ML Example
In linear regression, the formula for the best weights uses `X.T @ X` and `X.T @ y`. The transpose is also used to calculate covariance matrices and to match shapes between layers of a neural network.

#### Advantages
- Very fast, because no data is copied (it is a view).
- Often needed to make matrix shapes compatible.

#### Limitations
- Because it is a view, changing the transpose also changes the original array.
- It does nothing to a 1D array.

#### Key Point
Transpose swaps rows and columns: shape `(m, n)` becomes `(n, m)`. Use `.T`.

## Topic 14 – Basic Linear Algebra Operations

### Step 1 – Understand the Concept
NumPy has a `np.linalg` module for common linear algebra operations. The basic ones are:

| Function | What it gives |
|---|---|
| `np.linalg.det(A)` | Determinant of a square matrix |
| `np.linalg.inv(A)` | Inverse of a square matrix (`A @ inv(A)` = identity matrix) |
| `np.linalg.solve(A, b)` | Solution `x` of the equations `A x = b` |
| `np.linalg.norm(v)` | Length (magnitude) of a vector |
| `np.trace(A)` | Sum of the diagonal elements |
| `np.linalg.matrix_rank(A)` | Number of independent rows/columns |

An inverse exists only if the determinant is not zero.

### Step 2 – Demonstrate the Concept

#### Numerical Example
Solve the two equations:

2x + y = 5

x + 3y = 10

Written as `A x = b` with `A = [[2, 1], [1, 3]]` and `b = [5, 10]`. The answer is x = 1 and y = 3, because 2(1) + 3 = 5 and 1 + 3(3) = 10.

### Step 3 – Implement the Concept

```python
A = np.array([[2, 1],
              [1, 3]])
b = np.array([5, 10])

print("Determinant:", round(np.linalg.det(A), 2))

inverse = np.linalg.inv(A)
print("\nInverse:\n", np.round(inverse, 2))
print("A @ inverse (should be the identity matrix):\n", np.round(A @ inverse, 2))

solution = np.linalg.solve(A, b)
print("\nSolution of the equations [x, y]:", solution)

v = np.array([3, 4])
print("\nNorm of [3, 4]:", np.linalg.norm(v))
print("Trace of A    :", np.trace(A))
print("Rank of A     :", np.linalg.matrix_rank(A))

# ML-style: linear regression with the normal equation
# Data follows y = 1 + 2x. The first column of ones is for the intercept.
X = np.array([[1, 1],
              [1, 2],
              [1, 3],
              [1, 4]])
y = np.array([3, 5, 7, 9])
w = np.linalg.solve(X.T @ X, X.T @ y)
print("\nRegression weights [intercept, slope]:", np.round(w, 2))
```

#### What does this program do?
The program finds the determinant, inverse, solution of two equations, norm, trace and rank using `np.linalg`. It checks the inverse by multiplying it with the original matrix to get the identity matrix. At the end it finds the best line for a small dataset using the normal equation.

`np.linalg.solve` is used instead of calculating the inverse and multiplying, because it is faster and more accurate. The output shows a determinant of `5.0`, an inverse of `[[0.6 -0.2] [-0.2 0.4]]`, the identity matrix (a `-0.` is just a tiny rounding value), the solution `[1. 3.]`, a norm of `5.0`, a trace of `5` and a rank of `2`. The regression weights are `[1. 2.]`, which means intercept 1 and slope 2, matching the data `y = 1 + 2x`.

#### AI/ML Example
Linear regression can be solved directly with the formula `w = (XᵀX)⁻¹ Xᵀy`, as shown above. The norm is used for distances between data points (KNN, clustering) and in regularisation (L1/L2). Rank and determinant help to check if features are redundant.

#### Advantages
- Ready-made, fast and tested functions for common linear algebra tasks.
- Makes it possible to solve equations and models directly.

#### Limitations
- The inverse does not exist for a matrix with determinant 0 (a singular matrix), and NumPy gives a `LinAlgError`.
- For very large or nearly singular matrices, the results can be inaccurate.

#### Key Point
Use `np.linalg.solve(A, b)` to solve `A x = b`. Avoid using the inverse directly when `solve` can be used.

## Advantages of NumPy Mathematical Operations
- One function call replaces a loop, so the code is short and fast.
- Works on 1D, 2D and higher dimensional data with the `axis` argument.
- Gives all the statistics needed for data analysis and feature scaling.
- Matrix and linear algebra functions are the base of ML models.

## Limitations of NumPy Mathematical Operations
- Many functions need all the data to be in memory.
- Statistics functions do not ignore missing values unless the `nan` versions are used.
- Results are affected by outliers for mean, variance and standard deviation.
- Linear algebra functions can fail or be inaccurate for singular or very large matrices.

## Interview Questions

**1. What is the difference between mean and median? When should we use the median?**</br>

The mean is the sum of the values divided by their count, while the median is the middle value of the sorted data. The median should be used when the data has outliers or is skewed, because it is not affected by extreme values.

**2. What is the difference between population variance and sample variance in NumPy?**</br>

`np.var(data)` divides by n and gives the population variance. `np.var(data, ddof=1)` divides by n − 1 and gives the sample variance, which is used when the data is only a sample of a larger population.

**3. What is the difference between variance and standard deviation?**</br>

Standard deviation is the square root of the variance. Variance is in squared units, while standard deviation is in the same unit as the data, so it is easier to interpret.

**4. What does the `axis` parameter do?**</br>

`axis=0` performs the operation down the rows, giving one result for each column. `axis=1` performs it across the columns, giving one result for each row.

**5. What is the difference between `max` and `argmax`?**</br>

`max` returns the largest value, while `argmax` returns the index (position) of the largest value. In classification, `argmax` of the probabilities gives the predicted class.

**6. What is a percentile and how is it used to find outliers?**</br>

A percentile is the value below which a given percentage of the data lies. In the IQR rule, Q1 (25th percentile) and Q3 (75th percentile) are used, and values below Q1 − 1.5×IQR or above Q3 + 1.5×IQR are treated as outliers.

**7. What is the difference between `*`, `np.dot` and `@`?**</br>

`*` multiplies two arrays element by element. `np.dot` gives the dot product for 1D arrays and matrix multiplication for 2D arrays. `@` is the matrix multiplication operator and is the clearest choice for matrices.

**8. What is the shape rule for matrix multiplication?**</br>

The number of columns of the first matrix must equal the number of rows of the second. A `(m, n)` matrix multiplied by an `(n, p)` matrix gives an `(m, p)` result.

**9. Why is a dot product important in Machine Learning?**</br>

Each neuron calculates the dot product of its inputs and weights, and then adds a bias. Linear models also make predictions with a dot product, so it is the core calculation in many ML models.

**10. Why is `np.linalg.solve` preferred over `np.linalg.inv`?**</br>

`solve` finds the answer to `A x = b` directly, which is faster and numerically more accurate than calculating the inverse of `A` and multiplying it with `b`.
