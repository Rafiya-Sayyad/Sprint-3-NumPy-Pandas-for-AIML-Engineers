## Topic 1 – Rename Columns

### Step 1 – Understand the Concept
Renaming means changing the names of the columns (or the row labels). There are a few ways:

- df.rename(columns={"old": "new"}) changes only the columns that are given in the dictionary.
- df.rename(columns=function) applies a function to every column name, like str.upper.
- df.columns = [...] replaces all the names at once with a new list (the list must have the same number of names).
- df.columns.str.strip().str.lower().str.replace(" ", "_") cleans all the names together.
- df.rename(index={...}) changes the row labels in the same way.

rename() gives a new DataFrame, so the result has to be saved. If a name in the dictionary does not exist, rename() ignores it without an error (errors="raise" can be used to get an error).

### Step 2 – Demonstrate the Concept

#### Business Example
A report is going to the management, and the columns have names like "QTY" and " Unit Price " with extra spaces. Renaming them to clear and simple names makes the report look clean and easy to understand.

### Step 3 – Implement the Concept

```python
# A table with messy column names
raw = pd.DataFrame({
    "Order ID": [101, 102, 103],
    "Customer Name": ["Asha", "Ravi", "Meena"],
    " Unit Price ": [10, 250, 600],
    "QTY": [5, 2, 1]
})
print(raw.columns.tolist())

# Renaming only some columns with a dictionary
renamed = raw.rename(columns={"Order ID": "order_id", "QTY": "quantity"})
print()
print(renamed.columns.tolist())

# Applying a function to all the names
print()
print(raw.rename(columns=str.upper).columns.tolist())
```

```python
# Cleaning all the names together: remove spaces at the ends, lower case, spaces to underscores
clean = raw.copy()
clean.columns = clean.columns.str.strip().str.lower().str.replace(" ", "_")
print(clean.columns.tolist())
print()
print(clean)

# Renaming the row labels
print()
print(raw.rename(index={0: "first", 1: "second"}))

# A name that does not exist is ignored
print()
print(raw.rename(columns={"Wrong Name": "x"}).columns.tolist())
```

#### What does this program do?
The first cell prints the messy column names: Order ID, Customer Name, Unit Price (with spaces at both ends) and QTY. Then rename() with a dictionary changes only Order ID and QTY, so the output is order_id, Customer Name, " Unit Price " and quantity. The other names stay the same. After that, rename(columns=str.upper) changes all the names to capital letters.

In the second cell I made a copy of the table and cleaned all the names together with the string methods strip(), lower() and replace(). The new names are order_id, customer_name, unit_price and qty. The original table raw is not changed because I did this on a copy. Then rename(index=...) changes the first two row labels to first and second, and the third one stays 2. At the end, a name that does not exist is given to rename(), and the column names stay exactly the same, because no error is raised for a missing name.

#### AI/ML Example
Datasets from files often have column names with spaces, capital letters and extra spaces, like "Customer Name " with a space at the end. This causes KeyError mistakes when the name is typed without the space. So one of the first steps of data cleaning is to bring all the names to one clean style, such as lower case with underscores, so they are easy to type and the same across many datasets.

#### Advantages
- Makes the column names clear and consistent.
- The string methods can clean all the names in one line.
- Avoids mistakes caused by spaces and different capital letters.

#### Limitations
- A wrong name in the dictionary is ignored silently, so a spelling mistake can be missed.
- Assigning df.columns needs a list of the exact same length as the number of columns.

#### Key Point
rename(columns={"old": "new"}) renames chosen columns, and df.columns.str.strip().str.lower().str.replace(" ", "_") cleans all the names. Save the result because rename() returns a new table.

## Topic 2 – Drop Rows

### Step 1 – Understand the Concept
drop() removes rows (or columns) and gives back a new table. To remove rows, the labels of the rows are given:

- df.drop(index=[0, 1]) or df.drop([0, 1]) removes the rows with the labels 0 and 1.
- To remove rows by a condition, find the labels of the matching rows and drop them, or simply keep the rows that do not match with a filter.
- A label that does not exist gives a KeyError, unless errors="ignore" is used.
- After dropping, the row numbers have gaps, so reset_index(drop=True) is used to number them again.

The original table does not change, unless inplace=True is used (it is better to save the result in a variable instead).

### Step 2 – Demonstrate the Concept

#### Business Example
The sales report should only count orders that were really sold. The cancelled orders have to be removed from the orders table before the total sales are calculated.

### Step 3 – Implement the Concept

```python
print("Original shape:", orders.shape)

# Dropping rows by label
print("After dropping rows 0 and 1:", orders.drop(index=[0, 1]).shape)
print("After dropping row 3       :", orders.drop(3).shape)

# Dropping rows by a condition: first find the labels of the cancelled orders
cancelled = orders[orders["status"] == "cancelled"].index
print()
print("Cancelled order labels:", cancelled.tolist())

clean = orders.drop(index=cancelled)
print("Shape after removing cancelled orders:", clean.shape)
print("Remaining order ids:", clean["order_id"].tolist())

# Keeping the rows that are not cancelled gives the same table
print("Same as filtering?", clean.equals(orders[orders["status"] != "cancelled"]))
```

```python
# Fixing the row numbers
clean = clean.reset_index(drop=True)
print(clean.index.tolist())

# The original table is not changed
print()
print("Original shape is still:", orders.shape)

# A label that does not exist
try:
    orders.drop(index=[50])
except KeyError:
    print()
    print("KeyError: label 50 does not exist")

print("With errors='ignore':", orders.drop(index=[50], errors="ignore").shape)
```

#### What does this program do?
In the first cell, the original table has the shape (10, 6). Dropping the rows 0 and 1 gives (8, 6), and dropping only the row 3 gives (9, 6).

To remove the cancelled orders, I first find their row labels with a condition. The output is [3, 7], which are the orders 104 and 108. Dropping these labels gives a table of shape (8, 6), and the remaining order ids are 101, 102, 103, 105, 106, 107, 109 and 110. The last line of the cell shows that filtering for status not equal to cancelled gives exactly the same table, so the output is True.

In the second cell, reset_index(drop=True) numbers the rows from 0 to 7 again. The original table orders still has the shape (10, 6), which shows that drop() does not change it. Dropping the label 50, which does not exist, gives a KeyError, but with errors="ignore" it does nothing and the shape stays (10, 6).

#### AI/ML Example
Before training a model, rows that should not be used are removed, such as rows with wrong or impossible values (a negative age), rows of test records, or outliers. The same drop-by-condition idea is used. After the rows are removed, reset_index(drop=True) is used so that the row numbers are clean for the next steps.

#### Advantages
- Easy way to remove unwanted rows by label or by a condition.
- The original data stays safe because a new table is returned.

#### Limitations
- Dropped rows are lost from the new table, so the data size becomes smaller.
- A label that does not exist gives an error unless errors="ignore" is used.
- The row numbers have gaps unless they are reset.

#### Key Point
drop(index=[...]) removes rows by label, and it returns a new table. Use reset_index(drop=True) after dropping to fix the row numbers.

## Topic 3 – Drop Columns

### Step 1 – Understand the Concept
Columns can be removed in a few ways:

- df.drop(columns=["a", "b"]) removes the given columns and returns a new table. The same can be written as df.drop(["a", "b"], axis=1).
- del df["a"] removes a column from the table itself.
- df.pop("a") removes the column from the table itself and also gives it back as a Series.
- df[["x", "y"]] keeps only the chosen columns (the opposite way of thinking).

### Step 2 – Demonstrate the Concept

#### Business Example
A table is going to be shared with the customers, but the column order_id and the column status are for internal use only. They are removed before the table is sent.

### Step 3 – Implement the Concept

```python
print("Original columns:", orders.columns.tolist())

# drop with columns=
print()
print(orders.drop(columns=["order_id"]).columns.tolist())
print(orders.drop(columns=["order_id", "status"]).columns.tolist())

# the same with axis=1
print(orders.drop("qty", axis=1).columns.tolist())

# keeping only some columns
print()
print(orders[["customer", "product"]].shape)
```

```python
# del and pop change the table itself, so I use a copy
temp = orders.copy()

removed = temp.pop("status")
print("Popped column:", removed.head(3).tolist())
print("Columns now  :", temp.columns.tolist())

del temp["qty"]
print("After del    :", temp.columns.tolist())

print()
print("The original still has", orders.shape[1], "columns")
```

```python
# ML-style: features X and target y
y = orders["status"]
X = orders.drop(columns=["order_id", "status"])

print("Feature columns:", X.columns.tolist())
print("X shape:", X.shape, "| y shape:", y.shape)
```

#### What does this program do?
In the first cell, drop(columns=["order_id"]) gives the table without order_id, so 5 columns remain. Dropping both order_id and status leaves customer, product, price and qty. drop("qty", axis=1) does the same job as columns=, and selecting only customer and product gives a table of shape (10, 2).

In the second cell I used a copy, because pop() and del change the table itself. pop("status") removes the column and gives it back, so the first three values are delivered, delivered and pending. After del temp["qty"] the copy has the columns order_id, customer, product and price. The original table still has 6 columns.

In the third cell, the target y is the status column and the features X are all the columns except order_id and status. X has the shape (10, 4) with the columns customer, product, price and qty, and y has the shape (10,).

#### AI/ML Example
Before training, the columns that should not be used as features are dropped. These are the ID columns (they have a different value in every row, so they have no pattern), columns that have only one value, and the target column itself, which is taken out of the features and kept separately as y. This is the X and y split that is done in almost every ML project.

#### Advantages
- Easy to remove one or many columns in one line.
- drop() keeps the original table safe, and pop() gives the removed column back.

#### Limitations
- A column name that does not exist gives a KeyError.
- del and pop change the table itself, and the column cannot be recovered unless a copy was made.

#### Key Point
drop(columns=[...]) removes columns and returns a new table. del and pop change the table itself. Dropping the ID and target columns gives the features X.

## Topic 4 – Insert Columns

### Step 1 – Understand the Concept
A new column can be added in these ways:

- df["new"] = values : adds the column at the end. The values can be a calculation on other columns, one fixed value for every row, or a list with exactly one value per row.
- df.insert(position, "new", values) : adds the column at a chosen position (0 is the first position). The name must not already exist.
- np.where(condition, value_if_true, value_if_false) : creates a column that depends on a condition.

The length of a list must be equal to the number of rows, otherwise there is a ValueError.

### Step 2 – Demonstrate the Concept

#### Business Example
The orders table has the price and the quantity, but not the amount of each order. A new column total is added by multiplying the price and the quantity, so the shop can see how much each order is worth.

### Step 3 – Implement the Concept

```python
temp = orders.copy()

# A calculated column
temp["total"] = temp["price"] * temp["qty"]

# A column with one fixed value
temp["currency"] = "INR"

# A column at a chosen position (position 1 = second column)
temp.insert(1, "country", "India")

print(temp.columns.tolist())
print()
print(temp[["order_id", "price", "qty", "total"]].head(3))
print()
print("Sum of the total column:", temp["total"].sum())
```

```python
# A column that depends on a condition
temp["order_size"] = np.where(temp["qty"] >= 3, "bulk", "small")
print(temp["order_size"].value_counts())

# ML-style: a 0/1 feature
temp["is_bulk"] = (temp["qty"] >= 3).astype(int)
print()
print("Number of bulk orders:", temp["is_bulk"].sum())
```

```python
# Two mistakes that give errors
try:
    temp["bad"] = [1, 2, 3]
except ValueError as error:
    print("ValueError:", error)

try:
    temp.insert(0, "country", "India")
except ValueError as error:
    print("ValueError:", error)
```

#### What does this program do?
The first cell makes a copy of the orders table and adds three columns. total is price multiplied by qty, currency has the same value INR in every row, and insert(1, "country", "India") puts the country column at the second position. The printed column list is order_id, country, customer, product, price, qty, status, total and currency. The first three totals are 50, 500 and 600, and the sum of the total column is 4200.

In the second cell, np.where() creates the column order_size, which says bulk when the quantity is 3 or more and small otherwise. The count shows 6 small orders and 4 bulk orders. The column is_bulk does the same in the form of 1 and 0, so its sum is 4.

In the third cell, two mistakes are shown on purpose. A list of 3 values for a table with 10 rows gives a ValueError about the length of the values, and inserting a column with a name that already exists (country) gives a ValueError saying the column cannot be inserted because it already exists.

#### AI/ML Example
Creating new columns from the existing ones is called feature engineering, and it often improves a model more than changing the algorithm. Examples are total = price multiplied by qty, a bulk or not flag, the age of a customer from the date of birth, or a ratio of two columns. The 0/1 column made with astype(int) is the kind of feature that models can use directly.

#### Advantages
- Easy to create new information from existing columns, with no loop.
- insert() controls the position of the new column.
- np.where() makes columns from conditions in one line.

#### Limitations
- A list with the wrong length gives an error.
- insert() does not allow a name that already exists, and df["name"] = ... silently replaces an existing column with the same name.
- The new column is added to the table itself, so a copy is needed if the original should stay the same.

#### Key Point
df["new"] = ... adds a column at the end, insert() adds it at a position, and np.where() creates a column from a condition.

## Topic 5 – assign()

### Step 1 – Understand the Concept
assign() adds one or more new columns and gives back a new DataFrame. The original table does not change. The new columns are given as name=value pairs.

- The value can be a ready Series or a function (like a lambda). A function gets the table as it is at that moment, so a column made earlier in the same assign() can be used by a later one.
- Because it returns a new table, many assign() steps can be chained one after the other, with other methods like loc or sort_values.
- The new column names must be written like normal Python names (no spaces).

### Step 2 – Demonstrate the Concept

#### Business Example
A shop wants a report with the total, the GST (18 percent) and the grand total of every order, without changing the orders table itself. With assign() all three columns are made in one step, and the original table stays safe.

### Step 3 – Implement the Concept

```python
# One new column. The original table does not change.
report = orders.assign(total=orders["price"] * orders["qty"])
print("Report columns  :", report.columns.tolist())
print("Original columns:", orders.columns.tolist())

# A fixed value for all the rows
print()
print(orders.assign(country="India").shape)
```

```python
# Many columns in one chain. Later columns use the earlier ones.
summary = (
    orders
    .assign(
        total=lambda d: d["price"] * d["qty"],
        gst=lambda d: (d["total"] * 0.18).round(2),
        grand_total=lambda d: d["total"] + d["gst"],
        is_bulk=lambda d: (d["qty"] >= 3).astype(int)
    )
    .loc[:, ["order_id", "total", "gst", "grand_total", "is_bulk"]]
)

print(summary.head(3))
print()
print("Total grand amount:", round(summary["grand_total"].sum(), 2))
```

#### What does this program do?
In the first cell, assign(total=...) makes a new table called report with a total column added, so it has 7 columns, while the original orders table still has its 6 columns. Then assign(country="India") gives a table with the shape (10, 7).

In the second cell, a chain of four columns is created. A lambda function d receives the table and calculates each column. gst uses the total column that was created just before it, and grand_total uses both total and gst. After that, loc keeps only the columns that are needed. The first three rows show a total of 50, 500 and 600, with a GST of 9, 90 and 108 and a grand total of 59, 590 and 708, and a bulk flag of 1, 0 and 0. The total grand amount is 4956.0, which is 4200 plus 18 percent GST.

#### AI/ML Example
assign() is popular in feature engineering and data preparation pipelines. Many new features are created in a chain of steps, and the original raw table stays unchanged, so the same raw data can be used again. It also makes the steps easy to read from top to bottom.

#### Advantages
- The original table is never changed, so it is safe.
- Many columns can be created in a single readable step.
- Can be chained with other methods.

#### Limitations
- Column names cannot have spaces when used as name=value (df["new name"] = ... is needed for that).
- A new table is created every time, which uses more memory for very large data.

#### Key Point
assign(name=value, ...) returns a new table with the new columns. A function (lambda) can use the columns that were made before it.

## Topic 6 – apply()

### Step 1 – Understand the Concept
apply() runs a function on the data. It can be used in three ways:

- On a Series, series.apply(function) runs the function on every value.
- On a DataFrame with axis=1, df.apply(function, axis=1) runs the function once for every row (the function gets the whole row).
- On a DataFrame with axis=0 (default), df.apply(function) runs the function once for every column.

The function can be a normal def function, a lambda or a built-in function like len. Extra values can be passed to the function as keyword arguments.

apply() is flexible, but it runs like a loop, so it is slower than the vectorised operations (like df["a"] * df["b"]). It should be used when there is no ready-made vectorised way.

### Step 2 – Demonstrate the Concept

#### Business Example
The shop wants to put every product into a price group: low (below 100), medium (100 to 300) and high (above 300). There is no single built-in function for this rule, so a small function is written and applied to the price column.

### Step 3 – Implement the Concept

```python
def price_category(price):
    if price < 100:
        return "low"
    elif price <= 300:
        return "medium"
    else:
        return "high"

temp = orders.copy()
temp["price_category"] = temp["price"].apply(price_category)
print(temp[["product", "price", "price_category"]].head(4))
print()
print(temp["price_category"].value_counts())
```

```python
# apply on every row (axis=1): the function gets the whole row
row_total = orders.apply(lambda row: row["price"] * row["qty"], axis=1)
print(row_total.head(3).tolist())

# apply on every column (axis=0)
print()
print(orders[["price", "qty"]].apply(lambda col: col.max()))

# A built-in function, and a function with an extra argument
print()
print(orders["customer"].apply(len).head(3).tolist())

def add_gst(price, rate):
    return round(price * (1 + rate), 2)

print(orders["price"].apply(add_gst, rate=0.18).head(3).tolist())
```

#### What does this program do?
In the first cell, I wrote a function price_category() with the rule for the three groups and applied it to the price column of a copy of the table. The first four rows show the Pen (10) as low, the Book (250) as medium, the Bag (600) as high and the Bottle (120) as medium. The count of the groups is 4 medium, 3 low and 3 high orders.

In the second cell, apply(..., axis=1) runs the lambda once for each row and multiplies the price and the quantity, so the first three results are 50, 500 and 600. apply() on the two columns with axis=0 gives the maximum of each column, which is 600 for price and 10 for qty. apply(len) gives the length of each customer name (4, 4 and 5 for Asha, Ravi and Meena), and the last line passes the argument rate=0.18 to the function add_gst(), giving the prices with GST: 11.8, 295.0 and 708.0.

#### AI/ML Example
apply() is used for custom steps that are not available as ready-made functions, like turning a free text into a category by some rule, calculating a new feature from several columns of a row, or cleaning each value of a text column. For simple maths on columns, a vectorised operation is better because it is much faster on big data.

#### Advantages
- Very flexible: any Python function can be used.
- Works on values, rows or columns.
- Easy to read for custom rules.

#### Limitations
- Slower than vectorised operations, especially with axis=1 on large tables.
- Function mistakes are only found when the code runs.

#### Key Point
apply() runs a function on every value (Series), row (axis=1) or column (axis=0). Use it for custom rules, and prefer vectorised operations when possible.

# Topic 7 – map()

### Step 1 – Understand the Concept
map() works on a Series (one column) and changes each value using:

- a dictionary (a lookup table): every value is replaced by the matching value from the dictionary,
- or a function: the function is applied to every value.

If a value is not found in the dictionary, the result for that value is NaN. This is an important difference from replace(), which keeps the values that are not found as they were.

### Step 2 – Demonstrate the Concept

#### Business Example
The shop wants to know the sales by category, but the orders table has only the product names. A dictionary says which category each product belongs to, and map() uses it to create a category column.

### Step 3 – Implement the Concept

```python
category_map = {"Pen": "Stationery", "Book": "Stationery",
                "Bag": "Accessories", "Bottle": "Accessories"}

temp = orders.copy()
temp["category"] = temp["product"].map(category_map)
print(temp[["product", "category"]].head(4))
print()
print(temp["category"].value_counts())
```

```python
# ML-style: turning text into numbers (label encoding)
status_map = {"pending": 0, "delivered": 1, "cancelled": 2}
temp["status_code"] = temp["status"].map(status_map)
print(temp[["status", "status_code"]].head(5))

# A function instead of a dictionary
print()
print(orders["customer"].map(str.upper).head(3).tolist())

# Values that are not in the dictionary become NaN
partial = {"Pen": "Stationery"}
print()
print("Missing values after map:", orders["product"].map(partial).isna().sum())
```

#### What does this program do?
In the first cell, the dictionary category_map says that Pen and Book are Stationery and Bag and Bottle are Accessories. map() uses it on the product column and creates the new column category. The first four rows show Pen as Stationery, Book as Stationery, Bag as Accessories and Bottle as Accessories. The count is 6 Stationery orders (3 pens and 3 books) and 4 Accessories orders (3 bags and 1 bottle).

In the second cell, the status text is changed to numbers with a second dictionary (pending 0, delivered 1 and cancelled 2), so the first five rows give 1, 1, 0, 2 and 1. Then map(str.upper) uses a function and gives ASHA, RAVI and MEENA for the first three customers. At the end the dictionary has only one product, Pen, so the other 7 orders cannot be found and become NaN, and the count of missing values is 7.

#### AI/ML Example
Most ML models cannot use text, so categories are changed into numbers. map() with a dictionary is a simple way to do this for a column like status, gender or size (this is called label encoding). A lookup dictionary is also used to add information from a reference list, like the category of a product.

#### Advantages
- Very clear way to convert values with a lookup table.
- Fast for a single column.
- Works with a dictionary or a function.

#### Limitations
- Values that are not in the dictionary become NaN, which can hide mistakes, so the missing values should be checked.
- It works on one column (Series) only.

#### Key Point
map() changes the values of a column using a dictionary or a function. Values that are not in the dictionary become NaN.

## Topic 8 – replace()

### Step 1 – Understand the Concept
replace() changes values to other values. It can be used on a Series or on the whole DataFrame, and it keeps the values that are not mentioned as they are.

- replace(old, new) : one value to another value.
- replace([old1, old2], new) : many values to one value.
- replace({old1: new1, old2: new2}) : a dictionary of changes.
- On a DataFrame, a nested dictionary like {"column": {old: new}} changes values only in the given column.
- replace(pattern, new, regex=True) : changes text using a pattern (regular expression). Regular expressions are explained in the string notebook.

### Step 2 – Demonstrate the Concept

#### Business Example
A customer table stores the city as short codes (HYD, CHE, PUN and MUM). For a report, the codes should be changed to the full city names. A dictionary with the codes and the names is used in replace().

### Step 3 – Implement the Concept

```python
messy = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meena", "Kiran", "Sana"],
    "city": ["HYD", "CHE", "PUN", "HYD", "MUM"],
    "age": [25, -1, 29, -1, 36],
    "income": ["40000", "N/A", "58000", "?", "45000"]
})
print(messy)

# Codes to full names
print()
print(messy["city"].replace({"HYD": "Hyderabad", "CHE": "Chennai", "PUN": "Pune", "MUM": "Mumbai"}))
```

```python
# ML-style: placeholder values to NaN
print(messy["age"].replace(-1, np.nan))

print()
print(messy["income"].replace(["N/A", "?"], np.nan))

# A full clean up with a nested dictionary, then fixing the type of income
cleaned = messy.replace({"age": {-1: np.nan}, "income": {"N/A": np.nan, "?": np.nan}})
cleaned["income"] = pd.to_numeric(cleaned["income"])
print()
print(cleaned)
print()
print(cleaned.dtypes)
print()
print(cleaned.isna().sum())
```

```python
# Removing text with a regular expression
prices_text = pd.Series(["Rs 100", "Rs 250", "Rs 40"])
prices = pd.to_numeric(prices_text.replace(r"Rs ", "", regex=True))
print(prices.tolist())

# replace keeps the values that are not found, map makes them NaN
partial = {"Pen": "Pencil"}
print()
print("replace:", orders["product"].replace(partial).unique().tolist())
print("map    :", orders["product"].map(partial).unique().tolist())
```

#### What does this program do?
The first cell creates a small messy table with city codes, the value -1 used for a missing age, and the texts N/A and ? used for a missing income. The replace() with a dictionary changes the city codes to Hyderabad, Chennai, Pune, Hyderabad and Mumbai.

In the second cell, replace(-1, np.nan) changes the -1 values of age to NaN, so the ages become 25.0, NaN, 29.0, NaN and 36.0 (decimals, because a column with NaN is float). The same is done with the list ["N/A", "?"] for income. In the third part, a nested dictionary replaces the placeholders of both columns in one step, and then pd.to_numeric() changes the income column from text to numbers. The data types are then text for name and city, float64 for age and float64 for income, and isna().sum() shows 2 missing values in age and 2 in income, and 0 in name and city.

In the third cell, the regular expression removes the text "Rs " from the prices, and to_numeric() changes them to the numbers 100, 250 and 40. At the end, replace() and map() are compared with the dictionary {"Pen": "Pencil"}. replace() changes only the Pen and keeps all the other products, so the list is Pencil, Book, Bag and Bottle. map() changes Pen to Pencil but all the other products become NaN.

#### AI/ML Example
Real datasets use all kinds of placeholders for missing values, like -1, 999, N/A, ? or an empty text. Before modelling, these are replaced with NaN so that the missing values can be handled in the right way, for example filled or removed. Replace is also used to fix spelling variations and to remove symbols like currency signs and commas so that a text column can become numeric.

#### Advantages
- Fixes many values at once with a list or a dictionary.
- Keeps the values that are not mentioned.
- Works on a column, on many columns or on the whole table.

#### Limitations
- Matches the whole value only (the text "Rs 100" is not changed by replacing "Rs" unless regex=True is used).
- It is case-sensitive for text.
- A column may stay as text after the replacement, so its type may need to be fixed with to_numeric() or astype().

#### Key Point
replace() changes chosen values and keeps all the others. Replacing placeholders like -1 or N/A with np.nan is a common cleaning step.

## Topic 9 – Duplicate Handling

### Step 1 – Understand the Concept
Duplicate rows are rows that repeat the same data. Two methods handle them:

- duplicated() gives True for every row that is a repeat of an earlier row. Summing it counts the duplicates.
- drop_duplicates() removes the duplicate rows.

Useful options of both:
- subset=[...] : check only the given columns to decide if rows are duplicates, and not the whole row.
- keep="first" (default) : the first row is kept and the later repeats are duplicates. keep="last" keeps the last one. keep=False marks all the repeated rows as duplicates, including the first one, so all of them are removed by drop_duplicates().
- ignore_index=True (in drop_duplicates) : gives new row numbers 0, 1, 2 after the removal.

### Step 2 – Demonstrate the Concept

#### Business Example
A company collected the names and emails of customers from two forms, and some customers filled both forms. If the table is used as it is, these customers get the same mail twice. Removing the duplicates gives a clean mailing list.

### Step 3 – Implement the Concept

```python
contacts = pd.DataFrame({
    "name": ["Asha", "Ravi", "Asha", "Meena", "Ravi", "Kiran", "Meena"],
    "city": ["Hyderabad", "Chennai", "Hyderabad", "Pune", "Chennai", "Mumbai", "Pune"],
    "email": ["asha@mail.com", "ravi@mail.com", "asha@mail.com", "meena@mail.com",
              "ravi@new.com", "kiran@mail.com", "meena@mail.com"]
})
print(contacts)

# Finding the duplicates
print()
print(contacts.duplicated())
print()
print("Number of duplicate rows:", contacts.duplicated().sum())
print()
print(contacts[contacts.duplicated()])
```

```python
# Removing the duplicates
print("Original rows:", len(contacts))
print("After drop_duplicates()          :", len(contacts.drop_duplicates()))
print("After drop_duplicates(keep='last'):", len(contacts.drop_duplicates(keep="last")))
print("After drop_duplicates(keep=False) :", len(contacts.drop_duplicates(keep=False)))

# Checking only some columns with subset
print()
by_name_city = contacts.drop_duplicates(subset=["name", "city"])
print(by_name_city)
print()
print("Rows left:", len(by_name_city))

# Fresh row numbers
print()
print(contacts.drop_duplicates(ignore_index=True).index.tolist())
```

```python
# The orders table has no repeated full rows, but some customers ordered the same product twice
print("Full row duplicates:", orders.duplicated().sum())
print("Same customer and product:", orders.duplicated(subset=["customer", "product"]).sum())
```

#### What does this program do?
The first cell creates a contacts table with 7 rows. Asha (row 2) and Meena (row 6) are exact repeats of earlier rows, and Ravi (row 4) has the same name and city as row 1 but a different email. duplicated() gives False for all rows except rows 2 and 6, so the count is 2, and the last print shows these two rows.

In the second cell, drop_duplicates() removes the 2 repeated rows and leaves 5 rows (keeping the first of each). With keep="last" it also leaves 5 rows, but the earlier copies of Asha and Meena are removed instead of the later ones. With keep=False all the rows that are repeated, including the first copies, are removed, so only 3 rows remain (Ravi with the first email, Ravi with the new email and Kiran). With subset=["name", "city"], only the name and the city are checked, so Ravi's second row is also a duplicate, and 4 rows are left (Asha, Ravi, Meena and Kiran). With ignore_index=True the row numbers become 0 to 4.

In the third cell, the orders table has no fully repeated rows, so the count is 0. But when only the customer and the product are checked, there are 2 repeats: Ravi ordered the Book twice and Meena ordered the Bag twice.

#### AI/ML Example
Duplicate rows in a dataset can make a model look better than it is. If the same row is in both the training and the testing part, the model has already seen the test row, which is called data leakage. Duplicates also give extra weight to some samples. So duplicates are checked and removed before the data is split. The subset option is used when a few columns decide that two rows are the same record.

#### Advantages
- Easy to find and count repeated rows.
- The subset and keep options give control over what a duplicate means.
- Gives a clean dataset for reports and models.

#### Limitations
- Only exact matches are found, so "Asha" and "asha " (different spelling) are not duplicates unless the text is cleaned first.
- Removing duplicates loses data, and sometimes repeated rows are valid (the same customer buying the same thing again).
- keep=False can remove too much, because it removes all the copies.

#### Key Point
duplicated() finds repeated rows and drop_duplicates() removes them. Use subset to check some columns only and keep to choose which copy stays.

## Interview Questions

#### 1. How do you remove rows and columns in pandas? Does drop() change the original table?

Use df.drop(index=[...]) to remove rows by label and df.drop(columns=[...]) to remove columns (or axis=0 for rows and axis=1 for columns). drop() returns a new table and does not change the original, unless inplace=True is used, so the result should be saved in a variable. To remove rows by a condition, filter the table or drop the labels of the matching rows.

#### 2. What is the difference between df["new"] = ..., insert() and assign()?

df["new"] = ... adds the column at the end of the table itself. insert(position, name, values) adds it at a chosen position, also in the table itself, and gives an error if the name already exists. assign(new=...) returns a new table with the column added and leaves the original unchanged, and it can be chained.

#### 3. What is the difference between apply(), map() and replace()?

apply() runs any function on every value, row or column, and it is the most flexible one but slower. map() changes the values of a Series with a dictionary or a function, and the values that are not in the dictionary become NaN. replace() changes only the values that are mentioned and keeps all the others as they are.

#### 4. How do you find and remove duplicate rows?

duplicated() marks the repeated rows (and sum() counts them), and drop_duplicates() removes them. The subset option checks only some columns, and keep decides which copy stays: "first" (default), "last" or False (removes all the copies).

#### 5. How do you clean the column names of a DataFrame?

Use rename(columns={"old": "new"}) to change some names, or change all of them together with df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_"). This removes the extra spaces, makes the names lower case and replaces the spaces with underscores, so that the names are easy to type and do not cause a KeyError.
