## Topic 1 – head()

### Step 1 – Understand the Concept
head() shows the first rows of a DataFrame. By default it shows 5 rows, and head(n) shows the first n rows. It is the quickest way to see what the data looks like.

### Step 2 – Demonstrate the Concept

#### Business Example
A company gets a sales file with 5 lakh rows. Opening the whole file is not possible, so the analyst first looks at the first few rows to check whether the columns and values look correct.

### Step 3 – Implement the Concept

```python
# First 5 rows (default)
print(customers.head())

# First 3 rows
print()
print(customers.head(3))
```

#### What does this program do?
The first line prints the first 5 rows of the customers table. The second one uses head(3) to print only the first 3 rows. The output has the same 8 columns and the row numbers 0 to 4 in the first table (Asha to Sana) and 0 to 2 in the second table (Asha, Ravi and Meena).

head() is used because printing the full table is not practical for big data, and a few rows are enough to check the column names and the type of values.

#### AI/ML Example
After loading a dataset with read_csv(), the first thing most people do is df.head(). It shows whether the file was read properly, whether the header is correct, and what kind of values each feature has, before any cleaning or model building starts.

#### Advantages
- Very fast, even for a huge dataset.
- Easy way to check if the data was loaded correctly.

#### Limitations
- Shows only the top rows, so it can give a wrong idea if the file is sorted (for example by date or by city).
- It does not show anything about missing values or the full data.

#### Key Point
head() shows the first 5 rows by default and head(n) shows the first n rows.

## Topic 3 – sample()

### Step 1 – Understand the Concept
sample() picks random rows from a DataFrame.

- sample(n) gives n random rows.
- sample(frac=0.25) gives 25 percent of the rows.
- random_state is a seed number. With the same random_state the same rows are picked every time, so the result can be repeated.
- sample(frac=1) gives all the rows in a random order, which shuffles the table.

The row numbers (index) of the picked rows are kept from the original table.

### Step 2 – Demonstrate the Concept

#### Real-World Example
To check the quality of soup, we taste one spoon and not the whole pot, but the soup should be stirred first so the spoon is a fair picture. In the same way, a random sample shows a fair picture of a table, and not just the top part.

### Step 3 – Implement the Concept

```python
# 3 random rows
print(customers.sample(3, random_state=42))

# 25 percent of the rows
print()
print(customers.sample(frac=0.25, random_state=1))

# The same random_state gives the same rows
a = customers.sample(3, random_state=7)
b = customers.sample(3, random_state=7)
print()
print("Same rows with the same random_state?", a.equals(b))
```

```python
# Shuffling the whole table and fixing the row numbers
shuffled = customers.sample(frac=1, random_state=0).reset_index(drop=True)
print(shuffled.head(3))
print()
print("Rows before:", len(customers), "| rows after shuffle:", len(shuffled))
```

#### What does this program do?
The first cell picks 3 random rows with sample(3), and 25 percent of the rows with sample(frac=0.25), which is 3 rows out of 12. Because the original row numbers are kept, the index of the picked rows is not in order. At the end it takes two samples with the same random_state and checks them with equals(), and the output is True.

The second cell uses sample(frac=1) to take all the rows in a random order, which is a shuffle, and then reset_index(drop=True) gives fresh row numbers 0, 1, 2 and so on. The number of rows is the same (12) before and after.

I cannot say which exact rows will be picked, because they are random, but with random_state they will be the same every time the cell is run.

#### AI/ML Example
sample() is used to look at random rows of a large dataset, especially when the file is sorted. It is also used to take a small random part of a big dataset for quick testing of code, and to shuffle the data before it is split into training and testing parts. random_state is used so that the result can be repeated.

#### Advantages
- Gives a fair picture of the data, better than head() for sorted data.
- Can shuffle the data in one line.
- Results can be repeated with random_state.

#### Limitations
- A small sample can miss rare values or small groups.
- Without random_state the rows change every time the code is run.

#### Key Point
sample(n) or sample(frac=...) picks random rows. Use random_state to get the same rows again, and sample(frac=1) to shuffle.

## Topic 4 – info()

### Step 1 – Understand the Concept
info() prints a short summary of the DataFrame. It shows:

- the number of rows (entries) and the type of index,
- the number of columns,
- for every column: its name, the count of non-null (not missing) values and the data type,
- the count of each data type and the memory used.

The non-null count is the most useful part, because if a column has fewer non-null values than the number of rows, it has missing values. info() prints the summary by itself and returns nothing.

In the dtype column, text columns are shown as object (or str in newer versions of pandas).

### Step 2 – Demonstrate the Concept

#### Business Example
A manager gets a customer file and wants to know how complete the data is. info() shows how many values are filled in every column, so it is easy to see which columns have gaps.

### Step 3 – Implement the Concept

```python
customers.info()
```

```python
# Non-null count of every column (the same numbers as in info())
print(customers.count())

# Number of missing values = total rows - non-null count
print()
print(len(customers) - customers.count())
```

#### What does this program do?
customers.info() prints the summary of the table. The output should show 12 entries and 8 columns. Most columns have 12 non-null values, but Age and Income have only 11, because each of them has one missing value. The data types are int64 for CustomerID and Purchases, float64 for Age and Income, and object (or str) for Name, City, Gender and Segment.

Age and Income are float64 and not int64, because a column with a missing value (NaN) has to be stored as decimals. This is a good example of how info() helps to understand the data.

In the second cell, count() gives the same non-null numbers as a Series, and subtracting it from the number of rows gives the missing values in every column. This gives 1 for Age, 1 for Income and 0 for the other columns.

#### AI/ML Example
Before building a model, info() is used to decide what has to be cleaned. Columns with missing values need to be filled or dropped, and columns with a wrong data type (like numbers stored as text) need to be converted. It gives this picture in one command.

#### Advantages
- Shows rows, columns, data types, missing values and memory in one go.
- Quick way to find missing values and wrong data types.

#### Limitations
- Only shows counts. It does not show the values, the spread or the outliers.
- It prints the result and does not return a table that can be used in code (count() can be used for that).

#### Key Point
info() gives the structure of the data: columns, non-null counts, data types and memory. A non-null count smaller than the number of rows means missing values.

## Topic 5 – describe()

### Step 1 – Understand the Concept
describe() gives summary statistics of the data. For number columns it shows:

- count : number of non-missing values
- mean : average
- std : standard deviation
- min and max : smallest and largest values
- 25%, 50% and 75% : the quartiles (the 50% is the median)

By default only the number columns are shown. describe(include="all") also adds the text columns, and for a text column it shows count, unique (number of different values), top (the most common value) and freq (how many times the top value appears). The percentiles option can be used to ask for other percentiles.

### Step 2 – Demonstrate the Concept

#### Numerical Example
For the Purchases column the values are 3, 7, 5, 12, 4, 2, 6, 15, 8, 10, 1, 6. The sum is 79, so the mean is 79 / 12 which is about 6.58. The smallest value is 1 and the largest is 15. describe() calculates all of this together in one step.

### Step 3 – Implement the Concept

```python
# Summary of the number columns
print(customers.describe())
```

```python
# Only some columns, rounded
print(customers[["Age", "Income"]].describe().round(2))

# Asking for different percentiles
print()
print(customers["Income"].describe(percentiles=[0.1, 0.9]))

# A text column
print()
print(customers["City"].describe())

# All columns together
print()
print(customers.describe(include="all"))
```

#### What does this program do?
The first cell prints the statistics of the number columns (CustomerID, Age, Income and Purchases). The count row shows 12 for CustomerID and Purchases but 11 for Age and Income, because describe() ignores missing values. The mean of Age is about 33.27, the mean of Income is about 57.82 and the mean of Purchases is about 6.58. The minimum and maximum of Age are 25 and 45, of Income 38 and 80, and of Purchases 1 and 15.

CustomerID is also in the table, but its statistics have no meaning, because it is only an ID.

The second cell shows the statistics of two columns rounded to 2 decimals, and the Income statistics with the 10% and 90% percentiles added. For the text column City, describe() shows count 12, unique 4, top Hyderabad and freq 4. With include="all", the number columns and the text columns are shown in one table, and the places that do not apply (like the mean of a text column) are shown as NaN.

#### AI/ML Example
describe() is used in Exploratory Data Analysis to check the scale and the spread of the features. If one feature has values in the lakhs and another in single digits (like Income and Age), feature scaling is needed. A maximum that is far away from the 75% value can be a sign of outliers, and a minimum like a negative age shows wrong data.

#### Advantages
- Gives many important statistics in one command.
- Helps to find outliers, wrong values and the need for scaling.
- Works for number and text columns.

#### Limitations
- Shows only summary numbers. It cannot show the shape of the distribution (a graph is needed for that).
- Ignores missing values, so it does not tell how many are missing (count helps a little).
- By default it ignores text columns.

#### Key Point
describe() gives count, mean, std, min, quartiles and max for number columns. Use include="all" to add text columns.

## Topic 6 – shape

### Step 1 – Understand the Concept
shape is an attribute (written without brackets) that gives the size of the DataFrame as a tuple of (number of rows, number of columns). Related things are:

- len(df) : number of rows
- df.size : total number of values (rows × columns)
- df.ndim : number of dimensions, which is always 2 for a DataFrame

### Step 2 – Demonstrate the Concept

#### Business Example
A company has a table of its customers. The shape tells how many customers are there (rows) and how many details are stored about each customer (columns).

### Step 3 – Implement the Concept

```python
print("Shape :", customers.shape)

rows, cols = customers.shape
print("Rows  :", rows)
print("Columns:", cols)

print()
print("len(customers):", len(customers))
print("size          :", customers.size)
print("ndim          :", customers.ndim)
```

#### What does this program do?
The program prints the shape of the table, splits it into the number of rows and columns, and then prints len(), size and ndim. The output is (12, 8), then 12 rows and 8 columns, then len 12, size 96 (12 × 8) and ndim 2.

shape is written without brackets, because it is an attribute and not a method.

#### AI/ML Example
In an ML project, the shape of the data is checked after loading, after every cleaning step and after the train-test split. It tells the number of samples and the number of features, and it shows quickly if some rows were lost by mistake (for example, after removing missing values).

#### Advantages
- Instant, it does not calculate anything.
- Helps to check each step of the data preparation.

#### Limitations
- Tells only the size, nothing about the content.
- Does not show how many values are missing.

#### Key Point
shape gives (rows, columns). Remember it has no brackets.

## Topic 7 – columns

### Step 1 – Understand the Concept
columns is an attribute that gives the column names of a DataFrame. It is an Index object, and it can be turned into a normal Python list with list(df.columns) or df.columns.tolist(). Column names can be checked, looped over and used to choose the features of a model.

### Step 2 – Demonstrate the Concept

#### Business Example
A manager gets a customer table and first wants to know which details are stored in it, such as name, age, city and income. The column names answer this without reading the data.

### Step 3 – Implement the Concept

```python
print("Columns:", customers.columns)

print()
print("As a list:", list(customers.columns))
print("Number of columns:", len(customers.columns))
print("Is Income a column?", "Income" in customers.columns)

# Checking for extra spaces in the column names
print()
print("Names with extra spaces:", [c for c in customers.columns if c != c.strip()])

# Choosing the feature columns and the target column
target = "Segment"
feature_cols = [c for c in customers.columns if c not in ["CustomerID", "Name", target]]
print()
print("Features:", feature_cols)
print("Target  :", target)
```

#### What does this program do?
The program prints the column names, turns them into a list, counts them (8) and checks if Income is a column (True). It then checks for column names that have extra spaces at the start or the end, and the list is empty, which means all the names are clean.

At the end it creates a list of feature columns by leaving out the ID, the name and the target column. The output shows the features Age, City, Gender, Income and Purchases, and the target Segment.

#### AI/ML Example
In ML the columns are divided into features (inputs) and the target (the thing to predict). ID and name columns are usually removed because they do not help the model. Checking the column names also helps to find spelling problems and extra spaces, which are a common reason for KeyError.

#### Advantages
- Quick way to know what information the data has.
- Can be used in loops and list comprehensions to select columns.

#### Limitations
- Shows only the names and not the type or the content of the columns.
- Names can look the same but have hidden spaces or different capital letters.

#### Key Point
columns gives the column names. Use list(df.columns) to get them as a list.

## Topic 8 – dtypes

### Step 1 – Understand the Concept
dtypes is an attribute that gives the data type of every column.

- int64 : whole numbers
- float64 : decimal numbers (also used when a number column has missing values)
- object (or str in newer pandas) : text
- bool : True or False
- datetime64 : dates and times

dtypes.value_counts() counts how many columns there are of each type, and select_dtypes() picks the columns of a given type.

### Step 2 – Demonstrate the Concept

#### Business Example
In the customers table, Purchases has whole numbers, so it is int64. Income is in thousands and has a missing value, so it is float64. City is text. A report that needs to add up the incomes works only if the type is numeric, so the type is checked first.

### Step 3 – Implement the Concept

```python
print(customers.dtypes)

print()
print(customers.dtypes.value_counts())

# Number columns and other columns
numeric_cols = customers.select_dtypes(include="number").columns.tolist()
other_cols = customers.select_dtypes(exclude="number").columns.tolist()
print()
print("Number columns:", numeric_cols)
print("Other columns :", other_cols)

# The type of a single column
print()
print("Type of Age:", customers["Age"].dtype)
```

#### What does this program do?
The program prints the data type of each column, counts how many columns there are of each type, and uses select_dtypes() to list the number columns and the other columns.

The output shows int64 for CustomerID and Purchases, float64 for Age and Income, and object (or str) for Name, City, Gender and Segment. The count of types is 4 text columns, 2 int64 columns and 2 float64 columns. The number columns are CustomerID, Age, Income and Purchases, and the other columns are Name, City, Gender and Segment.

Age is float64 because it has a missing value. This is why it is important to check dtypes.

#### AI/ML Example
Most ML models work only with numbers. dtypes tells which columns are numeric and which are text, so the text columns (like City and Gender) can be encoded into numbers before training. select_dtypes() is used to apply different preparation steps to the two groups of columns.

#### Advantages
- Quick check of the type of every column.
- select_dtypes() makes it easy to separate number and text columns.

#### Limitations
- Pandas can guess the type wrongly (for example, numbers stored as text will show as object), so the values should be checked too.
- The same name can be shown differently in different pandas versions (object or str).

#### Key Point
dtypes shows the data type of each column. A number column with missing values becomes float64.

## Topic 9 – memory_usage()

### Step 1 – Understand the Concept
memory_usage() gives the memory used by every column, in bytes. It also includes the index as the first row.

By default, text columns are counted only roughly. memory_usage(deep=True) goes into the values of text columns and gives the real memory used, so it is the number to trust for text data.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Before travelling, we weigh the luggage to check that it is within the limit. In the same way, before working on a big dataset we check how much memory it needs, to be sure it fits in the computer.

### Step 3 – Implement the Concept

```python
# Memory of each column (bytes), the quick way
print(customers.memory_usage())

# The real memory, including the text values
print()
print(customers.memory_usage(deep=True))

# Total in kilobytes
total_kb = customers.memory_usage(deep=True).sum() / 1024
print()
print("Total memory:", round(total_kb, 2), "KB")
```

#### What does this program do?
The first cell prints the memory used by the index and by each column. The int64 and float64 columns take 96 bytes each (12 rows × 8 bytes). The second cell uses deep=True, and now the text columns (Name, City, Gender and Segment) show a larger memory, because the real size of the text values is counted. The last cell adds everything and shows the total in kilobytes.

The exact numbers for the index and the text columns depend on the pandas version and the computer, so the numbers on your system may be a little different from other systems.

#### AI/ML Example
Before loading or working on a large dataset, memory_usage() shows whether it fits in the RAM. If it is too big, the memory can be reduced by using smaller data types (for example int32 instead of int64, or float32 instead of float64) and the category type for text columns. This is explained in the performance optimisation notebook.

#### Advantages
- Shows which columns use the most memory.
- Helps to plan how to make the data smaller.

#### Limitations
- Without deep=True, the memory of text columns is shown too small.
- deep=True can be slow for very large text columns.

#### Key Point
memory_usage(deep=True) gives the real memory used by each column in bytes. Divide the sum by 1024 to get KB.

## Topic 10 – nunique()

### Step 1 – Understand the Concept
nunique() counts the number of different (unique) values. It can be used on the whole DataFrame, which gives one number for every column, or on a single column. By default, missing values are not counted. nunique(dropna=False) counts missing value as a value too.

Related method: unique() gives the actual different values of a column.

### Step 2 – Demonstrate the Concept

#### Business Example
A company wants to know in how many different cities its customers live. nunique() on the City column gives that number (4), and unique() gives the names of the cities.

### Step 3 – Implement the Concept

```python
# Unique values in every column
print(customers.nunique())

# One column
print()
print("Number of cities:", customers["City"].nunique())
print("The cities      :", customers["City"].unique())

# Counting missing values as a value too
print()
print("Age, ignoring missing :", customers["Age"].nunique())
print("Age, with missing     :", customers["Age"].nunique(dropna=False))
```

```python
# Columns with only one value (they carry no information)
constant_cols = [c for c in customers.columns if customers[c].nunique() <= 1]
print("Constant columns:", constant_cols)

# Columns where every value is different (like IDs)
id_like_cols = [c for c in customers.columns if customers[c].nunique() == len(customers)]
print("ID like columns :", id_like_cols)
```

#### What does this program do?
The first cell prints the number of unique values in every column. The output should be CustomerID 12, Name 12, Age 11, City 4, Gender 2, Income 11, Purchases 11 and Segment 3. Age and Income show 11 because the missing value is not counted, and Purchases shows 11 because the value 6 appears twice. For the City column, nunique() gives 4 and unique() gives Hyderabad, Chennai, Pune and Mumbai. With dropna=False, the Age count becomes 12 because the missing value is counted as one more value.

In the second cell, a list comprehension finds the columns that have only one value (none in this table, so the list is empty) and the columns where every row has a different value, which are CustomerID and Name.

#### AI/ML Example
nunique() is used to understand the columns before modelling. A column with only one value is useless for a model and can be removed. A column where every value is different, like an ID, should not be used as a feature. A text column with only a few unique values (like Gender or Segment) can be encoded easily, while one with many unique values needs more careful handling.

#### Advantages
- Quick way to separate categorical columns from continuous ones.
- Helps to find useless columns and ID columns.

#### Limitations
- Counts unique values only. It does not say how often each value appears (value_counts() does that).
- Ignores missing values unless dropna=False is used.

#### Key Point
nunique() counts different values, unique() lists them, and nunique(dropna=False) also counts missing values.

## Topic 11 – value_counts()

### Step 1 – Understand the Concept
value_counts() works on one column (a Series) and counts how many times each different value appears. The result is sorted from the most common value to the least common one.

Useful options:
- normalize=True : gives the proportion of each value instead of the count
- ascending=True : sorts from the least common value
- dropna=False : counts missing values also
- sort_index() can be added at the end to sort by the value names

### Step 2 – Demonstrate the Concept

#### Business Example
A company wants to know how many customers it has in every city, so that it can decide where to open a new store. value_counts() on the City column gives the count for each city.

### Step 3 – Implement the Concept

```python
# Number of customers in each city
print(customers["City"].value_counts())

# As proportions and as percentages
print()
print(customers["City"].value_counts(normalize=True))
print()
print((customers["City"].value_counts(normalize=True) * 100).round(1))

# Least common first, and sorted by city name
print()
print(customers["City"].value_counts(ascending=True))
print()
print(customers["City"].value_counts().sort_index())
```

```python
# Checking the target column (Segment) and Gender
print(customers["Segment"].value_counts())
print()
print(customers["Gender"].value_counts())
```

#### What does this program do?
The first cell counts the customers of each city. The output shows Hyderabad with 4 on top, then Chennai and Pune with 3 each, and Mumbai with 2 at the end. With normalize=True the counts change into proportions (Hyderabad is 0.33, Chennai and Pune are 0.25 and Mumbai is 0.17), and after multiplying by 100 and rounding they become the percentages 33.3, 25.0, 25.0 and 16.7. With ascending=True the least common city comes first, and sort_index() puts the cities in alphabetical order.

The second cell counts the Segment column, which gives Silver 5, Gold 4 and Bronze 3, and the Gender column, which gives 6 for each of F and M.

#### AI/ML Example
For a classification problem, value_counts() on the target column shows if the classes are balanced. If one class has very few rows compared to the others (for example 95% not fraud and 5% fraud), the data is imbalanced and special techniques are needed. It is also used to look at the categories of a feature and to find rare or wrongly spelled categories.

#### Advantages
- Shows the distribution of a column in one line.
- Helps to find imbalanced classes and wrong category names.
- Can show counts or percentages.

#### Limitations
- Works on one column at a time.
- For a number column with many different values it gives a very long output (it is meant for categories).

#### Key Point
value_counts() counts how often each value appears in a column. Use normalize=True for proportions.

## Interview Questions

#### 1. What is the difference between info() and describe()?

info() shows the structure of the data: the number of rows and columns, the data type and the non-null count of every column, and the memory used. describe() shows summary statistics like count, mean, standard deviation, min, quartiles and max. info() is used to find missing values and wrong types, and describe() is used to understand the values and find outliers.

#### 2. What is the difference between head() and sample()? What is random_state?

head() always gives the first rows, which may not be a fair picture if the data is sorted. sample() gives random rows. random_state is a seed number, and with the same value the same random rows are picked every time, so the result can be repeated.

#### 3. What is the difference between nunique(), unique() and value_counts()?

nunique() gives the number of different values, unique() gives the different values themselves, and value_counts() gives each different value along with how many times it appears.

#### 4. Why do we use deep=True in memory_usage()?

Without deep=True, pandas does not look inside text columns, so their memory is shown too small. With deep=True it counts the real size of the text values, so the numbers are correct.

#### 5. What is the difference between shape, size and len() of a DataFrame?

shape gives (rows, columns). len(df) gives only the number of rows. size gives the total number of values, which is rows multiplied by columns. For a table with 12 rows and 8 columns, shape is (12, 8), len is 12 and size is 96.
