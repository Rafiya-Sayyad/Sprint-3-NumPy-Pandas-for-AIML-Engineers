## Topic 1 – Random Numbers

### Step 1 – Understand the Concept
A random number is a number picked by chance, so it cannot be predicted. Computers actually use a mathematical formula to create numbers that look random. These are called pseudo-random numbers, and they are good enough for almost every data science task.

The most common functions are:

| Function | What it gives |
|---|---|
| `np.random.rand(shape)` | Random decimals between 0 (included) and 1 (not included), all equally likely |
| `np.random.randn(shape)` | Random decimals from the standard normal distribution (mean 0, standard deviation 1), so some values are negative |

### Step 2 – Demonstrate the Concept

#### Real-World Example
A bank sends an OTP (one-time password) to a customer. The digits should be unpredictable, so they are chosen at random. In the same way, a random decimal between 0 and 1 can be used to represent a "chance", for example 0.3 means a 30% chance.

### Step 3 – Implement the Concept

```python
one_d = np.random.rand(3)
two_d = np.random.rand(2, 3)
normal_values = np.random.randn(3)

print("rand(3)     :", one_d)
print("rand(2, 3)  :\n", two_d)
print("randn(3)    :", normal_values)

# Checking the range of rand
big = np.random.rand(1000)
print("\nSmallest of 1000 values:", big.min())
print("Largest of 1000 values :", big.max())

# ML-style: starting the weights of a neural network layer with small random numbers
weights = np.random.randn(3, 2) * 0.01
print("\nInitial weights:\n", weights)
```

#### What does this program do?
The program creates random numbers in different shapes. `rand` gives decimals between 0 and 1, and `randn` gives decimals around 0 which can also be negative. It then checks the smallest and largest of 1000 values from `rand`, and finally creates a 3×2 array of small random weights.

These functions are used because the shape can be given directly, so a whole array of random numbers is made in one line. The output will show values between 0 and 1 for `rand`, both positive and negative values for `randn`, and a smallest value close to 0 and a largest value close to 1 for the 1000 values. The numbers will be different each time the cell is run.

#### AI/ML Example
When a neural network is created, its weights must start as small random numbers (not all zeros, otherwise all neurons would learn the same thing). Multiplying `np.random.randn` by 0.01 is a simple way to do this, as shown above.

#### Advantages
- Creates random numbers of any shape in one line.
- Useful for starting model weights and for creating practice data.

#### Limitations
- The numbers are pseudo-random (made by a formula), so they are not suitable for security purposes such as real passwords or encryption keys.
- Results change on every run unless a seed is set.

#### Key Point
`rand` gives random decimals between 0 and 1. `randn` gives random decimals from the standard normal distribution.

## Topic 2 – Random Seed

### Step 1 – Understand the Concept
A random seed is a starting number for the random number generator. If the same seed is set, the generator gives exactly the same "random" numbers every time. This is called **reproducibility**.

`np.random.seed(42)` sets the seed. The number 42 has no special meaning, any whole number can be used. The seed has to be set again whenever we want to restart the same sequence.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Think of a teacher who shuffles a deck of cards and writes down the order so that the same shuffled deck can be created again later. The seed is like that written order: it lets us get the same "random" result again.

### Step 3 – Implement the Concept

```python
np.random.seed(42)
first = np.random.rand(3)

np.random.seed(42)          # same seed again
second = np.random.rand(3)

third = np.random.rand(3)   # no new seed, the sequence just continues

print("First  :", first)
print("Second :", second)
print("Third  :", third)
print("\nFirst and second are equal?", np.array_equal(first, second))
print("Second and third are equal?", np.array_equal(second, third))

# The newer way: a Generator object with its own seed
rng = np.random.default_rng(42)
print("\ndefault_rng(42):", rng.random(3))
rng = np.random.default_rng(42)
print("default_rng(42) again:", rng.random(3))
```

#### What does this program do?
The program sets the seed to 42, creates three random numbers, sets the same seed again and creates three more. It then creates another three numbers without resetting the seed. At the end it shows the newer `default_rng` way of doing the same thing.

The seed is used to make results repeatable. The output shows that the first and second sets are exactly the same (`[0.37454012 0.95071431 0.73199394]`), while the third set is different because the sequence continued. The `default_rng(42)` results are also the same both times (`[0.77395605 0.43887844 0.85859792]`).

#### AI/ML Example
In ML projects, a seed is set so that the train/test split, weight starting values and shuffling are the same on every run. This allows other people (and the reviewer) to get the same results and to compare two models fairly.

#### Advantages
- Makes experiments and results repeatable.
- Makes debugging easier, because the same numbers appear again.

#### Limitations
- A seed does not make the numbers truly random, it only makes them predictable.
- The seed must be set again to repeat the same sequence, and results can change if the order of random calls in the code changes.

#### Note
`np.random.seed()` is the classic way and is used in most tutorials, so I use it in this notebook. NumPy now recommends `np.random.default_rng(seed)`, which gives a separate generator object. Both work, but they give different numbers for the same seed.

#### Key Point
Same seed = same random numbers every time. It is used for reproducible results.

## Topic 3 – Random Integers

### Step 1 – Understand the Concept
`np.random.randint(low, high, size)` gives random whole numbers from `low` (included) up to `high` (**not included**). The `size` can be a number or a shape like `(2, 3)`. If only one number is given, it is treated as `high` and `low` is 0.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Rolling a die gives a whole number from 1 to 6. This is `np.random.randint(1, 7)`, because the high value 7 is not included.

### Step 3 – Implement the Concept

```python
np.random.seed(42)

# 10 rolls of a die
rolls = np.random.randint(1, 7, 10)
print("Dice rolls:", rolls)

values, counts = np.unique(rolls, return_counts=True)
print("Values :", values)
print("Counts :", counts)

# 2D array of random integers
print("\nrandint(0, 10, (2, 3)):\n", np.random.randint(0, 10, (2, 3)))

# ML-style: random class labels (0, 1 or 2) for 8 samples, useful for dummy data
labels = np.random.randint(0, 3, 8)
print("\nRandom labels:", labels)
```

#### What does this program do?
The program rolls a die 10 times with `randint(1, 7, 10)` and counts how many times each number came up using `np.unique` with `return_counts=True`. It then creates a 2×3 array of whole numbers from 0 to 9 and finally creates random class labels (0, 1 or 2) for 8 samples.

`randint` is used when we need whole numbers (not decimals). The output shows the 10 rolls `[4 5 3 5 5 2 3 3 3 5]`, with counts of 1, 4, 1 and 4 for the numbers 2, 3, 4 and 5 (the numbers 1 and 6 did not appear in these 10 rolls), a 2×3 array of whole numbers below 10, and 8 labels that are 0, 1 or 2.

#### AI/ML Example
When building or testing a classification pipeline, random labels like the ones above are used as dummy data to check that the code works before the real data is ready. `randint` is also used to pick random positions (indexes) in an array.

#### Advantages
- Easy way to create whole numbers in any range and shape.
- Useful for simulations and test data.

#### Limitations
- The `high` value is never included, which is a common source of mistakes.
- Numbers can repeat, so it cannot be used when every number must be different (use `permutation` for that).

#### Key Point
`randint(low, high, size)` gives whole numbers from `low` up to `high - 1`.

## Topic 4 – Random Choice

### Step 1 – Understand the Concept
`np.random.choice(a, size, replace, p)` picks random elements from an existing array.

- `a` – the array (or a number n, which means `0` to `n-1`) to pick from.
- `size` – how many elements to pick.
- `replace` – if `True` (default), the same element can be picked more than once. If `False`, each element can be picked only once.
- `p` – optional list of probabilities for each element, so some elements can be picked more often than others.

### Step 2 – Demonstrate the Concept

#### Business Example
A company has 8 customers with the IDs 101 to 108 and wants to call 3 of them for a satisfaction survey. Each customer should be called at most once, so the company picks 3 customers at random **without replacement**.

### Step 3 – Implement the Concept

```python
np.random.seed(42)

# Business: pick 3 customers for a survey, no repeats
customer_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108])
survey = np.random.choice(customer_ids, 3, replace=False)
print("Customers selected for the survey:", survey)

# Choice with probabilities: a biased coin that gives Heads 70% of the time
flips = np.random.choice(["Heads", "Tails"], 1000, p=[0.7, 0.3])
names, counts = np.unique(flips, return_counts=True)
print("\nBiased coin, 1000 flips:", {str(k): int(v) for k, v in zip(names, counts)})

# ML-style: bootstrap sample (pick with replacement from the data)
data = np.array([10, 20, 30, 40, 50])
bootstrap = np.random.choice(data, 5, replace=True)
print("\nBootstrap sample:", bootstrap)

# Asking for more items than available without replacement
try:
    np.random.choice(data, 6, replace=False)
except ValueError as error:
    print("\nError:", error)
```

#### What does this program do?
The program picks 3 customer IDs without repeats, flips a biased coin 1000 times using the `p` argument, creates a bootstrap sample by picking 5 values with replacement, and finally shows the error when more items are requested than exist (with `replace=False`).

`choice` is used because it picks from our own data, not only from numbers. The output shows 3 different customers (`[102 106 101]`), about 70% Heads and 30% Tails (712 and 288 here), a bootstrap sample where some values repeat and some are missing (`[50 10 50 10 20]`), and a `ValueError` saying a larger sample than the population cannot be taken without replacement.

#### AI/ML Example
Picking a random part of a dataset is called sampling, and it is used to create small samples for quick testing. Picking with replacement is called bootstrapping, and it is used in methods like Random Forest, where every tree is trained on a different bootstrap sample of the data.

#### Advantages
- Works on any array, including names and labels, and not just numbers.
- Supports probabilities and with/without replacement in one function.

#### Limitations
- With `replace=False`, the sample size cannot be larger than the data.
- For a very large dataset, `choice` with `replace=False` can be slower than picking positions in other ways.

#### Key Point
`choice` picks random elements from your own array. Use `replace=False` for no repeats and `p` for different probabilities.

## Topic 5 – Shuffle

### Step 1 – Understand the Concept
`np.random.shuffle(array)` rearranges the elements of an array in a random order. It changes the **original array itself (in place)** and returns `None`. For a 2D array, it shuffles only the rows (the first axis), and the values inside each row stay together.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Shuffling a music playlist or a deck of cards: all the songs or cards are still there, only their order changes.

### Step 3 – Implement the Concept

```python
np.random.seed(42)

arr = np.arange(1, 9)
print("Before shuffle:", arr)
result = np.random.shuffle(arr)
print("After shuffle :", arr)
print("Value returned by shuffle:", result)

# 2D array: only the rows are shuffled
matrix = np.arange(1, 11).reshape(5, 2)
print("\nMatrix before:\n", matrix)
np.random.shuffle(matrix)
print("Matrix after (rows moved, values in each row kept together):\n", matrix)
```

#### What does this program do?
The program shuffles a 1D array of the numbers 1 to 8 and prints the array before and after. It also prints what `shuffle` returns. Then it shuffles a 5×2 matrix to show that whole rows move together.

`shuffle` is used to change the order of data without changing the data. The output shows the array changed to `[2 6 1 8 3 5 4 7]`, and `shuffle` returned `None` because the change was made in the original array. In the matrix, rows like `[1 2]` and `[5 6]` are still together but appear in a new order.

#### AI/ML Example
Training data is often shuffled before it is split or before every training round (epoch), so that the model does not learn from the order of the data (for example, if all the samples of one class are at the top of the file).

#### Advantages
- Fast, because it works in place and does not create a new array.
- Rows stay intact in a 2D array.

#### Limitations
- It changes the original array, so the original order is lost unless a copy was made first.
- Shuffling features and labels in two separate calls would break the match between them (use `permutation` indexes for that, see Topic 6).

#### Key Point
`shuffle` rearranges the original array in place and returns `None`. In 2D arrays, only the rows move.

## Topic 6 – Permutation

### Step 1 – Understand the Concept
`np.random.permutation()` gives a randomly arranged version of the values.

- If a **number n** is given, it returns a shuffled array of `0` to `n-1`.
- If an **array** is given, it returns a **shuffled copy** and the original stays unchanged.

| shuffle() | permutation() |
|---|---|
| Changes the original array | Returns a new array, original is not changed |
| Returns `None` | Returns the shuffled array |
| Takes only an array | Takes an array or a number |

### Step 2 – Demonstrate the Concept

#### Real-World Example
A teacher wants to decide a random order for 5 students to present. `np.random.permutation(5)` gives a random order of the numbers 0 to 4, and each student appears exactly once.

### Step 3 – Implement the Concept

```python
np.random.seed(42)

print("permutation(5):", np.random.permutation(5))

original = np.array([10, 20, 30, 40, 50])
shuffled_copy = np.random.permutation(original)
print("\nShuffled copy:", shuffled_copy)
print("Original      :", original)

# ML-style: shuffle features X and labels y together using the same indexes
X = np.arange(1, 11).reshape(5, 2)
y = np.array([0, 1, 0, 1, 1])
idx = np.random.permutation(len(X))
print("\nShuffled indexes:", idx)
print("X shuffled:\n", X[idx])
print("y shuffled:", y[idx])

# ML-style: 80/20 train-test split
np.random.seed(42)
idx = np.random.permutation(10)
split = int(0.8 * 10)
train_idx, test_idx = idx[:split], idx[split:]
print("\nTrain indexes:", train_idx)
print("Test indexes :", test_idx)
```

#### What does this program do?
The program creates a random order of the numbers 0 to 4, makes a shuffled copy of an array without changing the original, and then shuffles `X` and `y` together by using one set of shuffled indexes for both. At the end it makes an 80/20 train-test split from shuffled indexes.

Using the same indexes for `X` and `y` is important because every row of `X` must stay matched with its own label. The output shows the order `[1 4 2 0 3]`, a shuffled copy `[40 20 30 10 50]` with the original still `[10 20 30 40 50]`, and the same index order (`[1 0 3 4 2]`) applied to both `X` and `y`. The split gives 8 training indexes `[8 1 5 0 7 2 9 4]` and 2 test indexes `[3 6]`.

#### AI/ML Example
This is exactly how a train-test split is done: shuffle the row numbers, take the first 80% for training and the rest for testing. The same idea is used inside Scikit-learn's `train_test_split`.

#### Advantages
- Keeps the original data safe.
- Gives indexes that can be used to shuffle several arrays in the same way.

#### Limitations
- Creates a new array, so it uses extra memory for very large data.

#### Key Point
`permutation` returns a shuffled copy (or a shuffled range of numbers) and leaves the original unchanged. Use shuffled indexes to keep `X` and `y` matched.

## Topic 7 – Random Distributions

### Step 1 – Understand the Concept
A distribution describes how likely each value is. NumPy can create random numbers that follow common distributions:

| Function | Distribution | Typical use |
|---|---|---|
| `np.random.uniform(low, high, size)` | Uniform: every value in the range is equally likely | Random values in a range |
| `np.random.normal(mean, std, size)` | Normal (bell curve): most values are near the mean | Heights, marks, measurement errors |
| `np.random.binomial(n, p, size)` | Binomial: number of successes in `n` tries, each with probability `p` | Coin tosses, yes/no outcomes |
| `np.random.poisson(lam, size)` | Poisson: number of events in a fixed time, with an average of `lam` | Customers per hour, calls per minute |
| `np.random.exponential(scale, size)` | Exponential: waiting time between events, with average `scale` | Time between arrivals |

With more samples, the average of the random numbers gets closer to the true mean of the distribution.

### Step 2 – Demonstrate the Concept

#### Real-World Example
The heights of adults in a city usually follow a normal distribution. Suppose the average height is 165 cm and the standard deviation is 7 cm. Most people will be close to 165 cm, and very short or very tall people will be rare. `np.random.normal(165, 7, 1000)` creates 1000 such heights.

### Step 3 – Implement the Concept

```python
np.random.seed(42)

# Uniform
print("uniform(10, 20, 5):", np.random.uniform(10, 20, 5).round(2))

# Normal: 1000 heights
heights = np.random.normal(165, 7, 1000)
print("\nHeights -> mean:", heights.mean().round(2), "| std:", heights.std().round(2))
print("Shortest:", heights.min().round(1), "| Tallest:", heights.max().round(1))

# More samples -> mean gets closer to 165
for n in [10, 1000, 100000]:
    print("Mean of", n, "samples:", np.random.normal(165, 7, n).mean().round(3))

# Binomial: number of Heads in 10 tosses, repeated 10 times
print("\nbinomial(10, 0.5, 10):", np.random.binomial(10, 0.5, 10))

# Poisson: customers arriving per hour, average 5
arrivals = np.random.poisson(5, 10)
print("poisson(5, 10)       :", arrivals)
print("Mean of 1000 Poisson values:", np.random.poisson(5, 1000).mean().round(2))

# Exponential: waiting times, average 2 minutes
print("exponential(2, 5)    :", np.random.exponential(2, 5).round(2))

# ML-style: creating a practice dataset  y = 1 + 2x + noise
x = np.random.uniform(0, 10, 50)
noise = np.random.normal(0, 1, 50)
y = 1 + 2 * x + noise

X = np.column_stack([np.ones(50), x])           # first column of ones for the intercept
w = np.linalg.solve(X.T @ X, X.T @ y)           # normal equation from Notebook 2
print("\nWeights found from the noisy data [intercept, slope]:", w.round(2))
```

#### What does this program do?
The program creates random numbers from the uniform, normal, binomial, Poisson and exponential distributions. It checks the mean and standard deviation of 1000 normal heights, and shows that the mean gets closer to 165 as the number of samples grows. At the end it builds a practice dataset using the line `y = 1 + 2x` plus normal noise, and recovers the line with the normal equation from Notebook 2.

Each function is used to match the kind of data we want to create. The output shows 5 values between 10 and 20 for `uniform`, heights with a mean of about `165.18` and a standard deviation of `7.0` (shortest `140.6`, tallest `188.6`), and sample means of `166.897`, `164.937` and `164.991` for 10, 1000 and 100000 samples, so the mean gets closer to 165 with more samples. `binomial` gives numbers from 0 to 10 (the number of Heads), `poisson` gives whole numbers around 5 with a mean of `5.07` over 1000 values, and `exponential` gives positive decimals with an average of about 2. The final weights are `[1.29 1.95]`, which are close to the true intercept 1 and slope 2. They are not exact because of the random noise.

#### AI/ML Example
Random distributions are used to create synthetic (practice) datasets, as shown above, to test models when real data is not available. The normal distribution is also used to start neural network weights, and to add noise to data. Understanding distributions is the base for the sampling and probability ideas from Sprint 2.

#### Advantages
- Can create data that behaves like real-world data.
- Makes it possible to test algorithms when real data is not available.

#### Limitations
- Real data is rarely a perfect normal, binomial or Poisson distribution, so synthetic data is only an approximation.
- Small samples can look very different from the true distribution (as seen with 10 samples above).

#### Key Point
Pick the distribution that matches the data: `uniform` for equal chances, `normal` for bell-shaped data, `binomial` for success counts, `poisson` for event counts and `exponential` for waiting times.

## Advantages of the NumPy Random Module
- Creates random numbers of any shape in one line.
- Gives many ready-made distributions.
- Supports reproducible results through seeds.
- Covers the common needs of ML: shuffling, sampling, splitting and weight initialisation.

## Limitations of the NumPy Random Module
- The numbers are pseudo-random, so they are not secure.
- Results depend on the seed and on the order of random calls.
- The classic `np.random.seed` uses one global generator, which can cause hidden effects in bigger projects (the `default_rng` generator avoids this).
- Synthetic random data is only an approximation of real data.

## Interview Questions

**1. What is a random seed and why is it used?**</br>

A seed is the starting number of the random number generator. Setting the same seed gives the same random numbers every time, which makes experiments reproducible and easier to debug and compare.

**2. What is the difference between `np.random.rand` and `np.random.randn`?**</br>

`rand` gives decimals between 0 and 1 that are all equally likely (uniform distribution). `randn` gives decimals from the standard normal distribution (mean 0, standard deviation 1), so values can be negative and are mostly close to 0.

**3. Is the high value included in `np.random.randint`?**</br>

No. `randint(1, 7)` gives whole numbers from 1 to 6. The high value is not included.

**4. What is the difference between `shuffle` and `permutation`?**</br>

`shuffle` rearranges the original array in place and returns `None`. `permutation` returns a new shuffled array (or a shuffled range if a number is given) and leaves the original unchanged.

**5. How can you shuffle features `X` and labels `y` together?**</br>

Create shuffled indexes with `idx = np.random.permutation(len(X))` and then use `X[idx]` and `y[idx]`. The same indexes keep every row of `X` matched with its own label.

**6. What does the `replace` parameter do in `np.random.choice`?**</br>

With `replace=True` the same element can be picked more than once (sampling with replacement). With `replace=False` every element can be picked at most once, so the sample size cannot be larger than the data.

**7. What is bootstrapping and where is it used?**</br>

Bootstrapping means creating a sample by picking from the data with replacement. It is used in methods like Random Forest, where every tree is trained on a different bootstrap sample.

**8. Which distribution would you use for the number of customers arriving per hour?**</br>

The Poisson distribution (`np.random.poisson`), because it models the number of events in a fixed time when events happen at a steady average rate.

**9. How do you do a train-test split with NumPy?**</br>

Shuffle the row numbers with `np.random.permutation(n)`, take the first 80% of the indexes for training and the remaining 20% for testing, and use those indexes to select rows from `X` and `y`.

**10. What is the difference between `np.random.seed()` and `np.random.default_rng()`?**</br>

`np.random.seed()` sets the seed of one global (classic) generator. `default_rng(seed)` creates a separate generator object with its own seed, which is the newer method recommended by NumPy because it avoids shared global state. The two give different numbers for the same seed.
