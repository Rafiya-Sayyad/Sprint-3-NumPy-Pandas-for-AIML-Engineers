## Topic 1 – What is NumPy?

### Step 1 – Understand the Concept
NumPy stands for Numerical Python. It is a Python library used to work with large collections of numbers.

Its main object is the `ndarray` (N-dimensional array). NumPy also gives many ready-made functions for maths, statistics and linear algebra that work on a whole array at once.

#### Advantages
- Makes numerical work short and easy to write.
- Works as the base for most data science libraries.

#### Limitations

#### Key Point
NumPy = a Python library for fast numerical computing, built around the `ndarray`.

## Topic 2 – Why NumPy?

### Step 1 – Understand the Concept
Python lists can store anything, but they are slow for numerical work. NumPy is used because:

- It is **fast**, because the operations run in compiled C code and work on the whole array at once. This is called vectorization.
- It uses **less memory**, because all elements have the same type and are stored together in one block of memory.
- It has **hundreds of built-in functions** for maths and statistics.
- Libraries like Pandas, Scikit-learn and PyTorch work with NumPy arrays.

### Step 2 – Demonstrate the Concept

#### Real-World Example
A teacher wants to add 5 bonus marks to every student. With a Python list, she updates one student at a time. With NumPy, it is like changing the whole marksheet in one step.

Below, I compare how long a Python list and a NumPy array take to square one million numbers.

### Step 3 – Implement the Concept

```python
import time
import sys

n = 1_000_000

# Python list
py_list = list(range(n))
start = time.perf_counter()
squared_list = [x ** 2 for x in py_list]
list_time = time.perf_counter() - start

# NumPy array
np_array = np.arange(n)
start = time.perf_counter()
squared_array = np_array ** 2
numpy_time = time.perf_counter() - start

print(f"Python list time : {list_time:.4f} seconds")
print(f"NumPy array time : {numpy_time:.4f} seconds")
print(f"NumPy was about {list_time / numpy_time:.0f}x faster")

# Memory used by 1000 numbers
small_list = list(range(1000))
small_array = np.arange(1000)
list_memory = sys.getsizeof(small_list) + sum(sys.getsizeof(i) for i in small_list)
print("\nMemory of list  (bytes):", list_memory)
print("Memory of array (bytes):", small_array.nbytes)
```

#### What does this program do?
The program squares one million numbers twice: once with a list comprehension and once with `np_array ** 2`. It measures both times using `time.perf_counter()`. It then compares the memory used by 1000 numbers stored in a list and in an array.

The output shows that NumPy is many times faster and uses much less memory. The exact speed difference changes from computer to computer.

#### AI/ML Example
Training data can have millions of rows. Looping over every row in Python would be too slow, so ML code uses vectorized NumPy operations for tasks like scaling features or calculating errors.

#### Advantages
- Much faster than Python loops for numerical work.
- Uses less memory than lists.

#### Limitations
- All elements must be of the same type.
- The size of an array is fixed after it is created.

#### Key Point
NumPy is preferred over lists for numbers because it is faster and more memory-efficient.

## Topic 3 – NumPy Arrays (ndarray)

### Step 1 – Understand the Concept
The `ndarray` is the main data structure of NumPy. It is a grid of values where:

- all values have the **same data type**,
- every value has a **fixed position** (index),
- it can have **any number of dimensions** (1D, 2D, 3D and more).

Every array has attributes such as `ndim`, `shape`, `size` and `dtype` that describe it.

### Step 2 – Demonstrate the Concept

#### Business Example
A company records the revenue (in lakhs) of 3 products over 4 quarters. This is a table with 3 rows and 4 columns, which is a 2D array.

| | Q1 | Q2 | Q3 | Q4 |
|---|---|---|---|---|
| Product 1 | 10 | 12 | 14 | 16 |
| Product 2 | 20 | 22 | 21 | 25 |
| Product 3 | 5 | 6 | 8 | 9 |

### Step 3 – Implement the Concept

```python
revenue = np.array([[10, 12, 14, 16],
                    [20, 22, 21, 25],
                    [ 5,  6,  8,  9]])

print(revenue)
print("\nType      :", type(revenue))
print("Dimensions:", revenue.ndim)
print("Shape     :", revenue.shape)
print("Data type :", revenue.dtype)
```

#### What does this program do?
The program creates a 2D array from a list of lists and prints its type, number of dimensions, shape and data type.

These attributes are printed because they tell us the structure of the data, which is the first thing to check before working on it. The output shows the 3×4 grid, `ndim = 2`, `shape = (3, 4)` and an integer data type (usually `int64`).

#### AI/ML Example
A colour image is stored as a 3D array of height × width × 3 colour channels. The weights of a neural network layer are stored as 2D arrays. Both are `ndarray` objects.

#### Advantages
- One structure can store 1D, 2D, 3D or higher-dimensional data.
- Built-in attributes make it easy to inspect the data.

#### Limitations
- Elements must be of one data type.
- It does not have column names or labels like a Pandas DataFrame.

#### Key Point
ndarray = a fixed-size grid of same-type values with any number of dimensions.

## Topic 4 – Creating Arrays

### Step 1 – Understand the Concept
NumPy gives several ways to create arrays.

| Function | What it creates |
|---|---|
| `np.array(list)` | Array from a list or nested list |
| `np.zeros(shape)` | Array filled with 0 |
| `np.ones(shape)` | Array filled with 1 |
| `np.full(shape, value)` | Array filled with a chosen value |
| `np.arange(start, stop, step)` | Values with a fixed step (stop is not included) |
| `np.linspace(start, stop, num)` | A fixed number of equally spaced values (stop is included) |
| `np.eye(n)` | Identity matrix |

### Step 2 – Demonstrate the Concept

#### Numerical Example
- `np.arange(0, 20, 5)` starts at 0, jumps by 5 and stops before 20, so the result is `0, 5, 10, 15`.
- `np.linspace(0, 1, 5)` gives 5 equally spaced values from 0 to 1, so the result is `0, 0.25, 0.5, 0.75, 1`.

### Step 3 – Implement the Concept

```python
a = np.array([1, 2, 3, 4, 5])      # from a list
b = np.zeros((2, 3))               # 2 rows, 3 columns of 0
c = np.ones((2, 3))                # 2 rows, 3 columns of 1
d = np.full((2, 2), 7)             # 2 x 2 filled with 7
e = np.arange(0, 20, 5)            # 0, 5, 10, 15
f = np.linspace(0, 1, 5)           # 5 values from 0 to 1
g = np.eye(3)                      # 3 x 3 identity matrix

print("np.array :", a)
print("zeros    :\n", b)
print("ones     :\n", c)
print("full     :\n", d)
print("arange   :", e)
print("linspace :", f)
print("eye      :\n", g)
```

#### What does this program do?
The program creates seven arrays using seven different functions and prints them.

Different functions are used for different needs: `zeros`, `ones` and `full` create placeholders, `arange` and `linspace` create sequences, and `eye` creates an identity matrix. In the output, `arange` stops before 20, while `linspace` includes the end value 1.

#### AI/ML Example
When a neural network is built, its biases are often started with `np.zeros`. `np.linspace` is used to create a range of values to try for a setting like the learning rate, and `np.eye` is used in linear algebra.

#### Advantages
- Many ready-made functions, so there is no need to type values by hand.
- Easy to create large arrays in one line.

#### Limitations
- `arange` with decimal steps can give unexpected lengths because of floating-point errors, so `linspace` is safer for decimals.

#### Key Point
`arange` works with a step size, while `linspace` works with the number of points.

## Topic 5 – Dimensions

### Step 1 – Understand the Concept
The number of dimensions (axes) of an array is given by `ndim`.

| ndim | Name | Meaning |
|---|---|---|
| 0 | Scalar | a single number |
| 1 | Vector | a list of numbers |
| 2 | Matrix | rows and columns |
| 3 | Tensor | a stack of matrices |

### Step 2 – Demonstrate the Concept

#### Real-World Example
- One temperature reading is 0D.
- Temperatures of 7 days form a 1D array.
- Temperatures of 3 cities for 7 days form a 2D array.
- The same data for 12 months forms a 3D array.

### Step 3 – Implement the Concept

```python
scalar = np.array(42)
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2, 3], [4, 5, 6]])
tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("Scalar ndim:", scalar.ndim)
print("Vector ndim:", vector.ndim)
print("Matrix ndim:", matrix.ndim)
print("Tensor ndim:", tensor.ndim)

# Batch of 32 colour images, each 64 x 64 pixels with 3 colour channels
images = np.zeros((32, 64, 64, 3))
print("\nImage batch shape:", images.shape, "| ndim:", images.ndim)

# Forcing a minimum number of dimensions
five_d = np.array([1, 2, 3], ndmin=5)
print("ndmin=5 ->", five_d, "| ndim:", five_d.ndim)
```

#### What does this program do?
The program creates arrays with 0, 1, 2 and 3 dimensions and prints `ndim` for each. It then creates an array that looks like a batch of 32 colour images, and uses `ndmin` to force an array to have 5 dimensions.

`ndim` is used to know how deeply the data is nested. The output is `0, 1, 2, 3` for the first four arrays, `ndim = 4` for the image batch and `ndim = 5` for the last one.

#### AI/ML Example
One sample's features are 1D, a whole dataset is 2D, a colour image is 3D and a batch of colour images is 4D. Many model errors happen because data is given with the wrong number of dimensions, so `ndim` is checked often.

#### Key Point
ndim = the number of axes of an array (0D scalar, 1D vector, 2D matrix, 3D+ tensor).

## Topic 6 – Shape

### Step 1 – Understand the Concept
`shape` is a tuple that tells how many elements are there along each dimension. For a 2D array, `(5, 3)` means 5 rows and 3 columns.

The length of the shape tuple is equal to `ndim`.

### Step 2 – Demonstrate the Concept

#### Business Example
A company has data of 500 customers and records 8 details of each one. The table has 500 rows and 8 columns, so its shape is `(500, 8)`.

### Step 3 – Implement the Concept

```python
# 5 customers, 3 features each: age, income (thousands), purchases
customers = np.array([[25, 40, 3],
                      [32, 55, 7],
                      [47, 80, 12],
                      [51, 62, 5],
                      [29, 45, 4]])

print("Shape             :", customers.shape)
print("Rows (samples)    :", customers.shape[0])
print("Columns (features):", customers.shape[1])
```

#### What does this program do?
The program creates a small customer table and prints its shape. It then reads the number of rows and columns from the shape tuple using index `[0]` and `[1]`.

Shape is used because it is the quickest way to understand the size of the data. The output is `(5, 3)`, `5` and `3`.

#### AI/ML Example
In almost every ML project, `X.shape` is checked to confirm that the data is in the form `(number of samples, number of features)`. Many errors in neural networks are "shape mismatch" errors.

#### Key Point
shape = the size of the array along each dimension, written as a tuple.

## Topic 7 – Size

### Step 1 – Understand the Concept
`size` is the total number of elements in an array. It is the product of all values in the shape.

Two related attributes are:
- `itemsize` – memory used by one element (in bytes).
- `nbytes` – total memory used by the array (`size × itemsize`).

### Step 2 – Demonstrate the Concept

#### Real-World Example
A cinema hall has 10 rows and 15 seats in each row. The shape is `(10, 15)` and the size is 10 × 15 = 150, which is the total number of seats.

### Step 3 – Implement the Concept

```python
print("Shape    :", customers.shape)
print("Size     :", customers.size)
print("Itemsize :", customers.itemsize, "bytes per element")
print("Nbytes   :", customers.nbytes, "bytes in total")

# Estimating memory of a big dataset without creating it:
# 1,000,000 rows x 100 features stored as float64
rows, cols = 1_000_000, 100
memory_mb = rows * cols * np.dtype("float64").itemsize / 1_000_000
print("\nEstimated memory of a 1,000,000 x 100 float64 dataset:", memory_mb, "MB")
```

#### What does this program do?
The program prints the size, item size and total memory of the `customers` array from the previous topic. It then calculates how much memory a large dataset of 1,000,000 rows and 100 features would need, without actually creating it.

The output is `size = 15` (5 × 3), `itemsize = 8` bytes, `nbytes = 120`, and an estimate of `800.0 MB` for the large dataset.

#### AI/ML Example
Before loading a large dataset or a batch of images, an engineer can estimate its memory using `size` and `itemsize`. If it is too big, a smaller data type like `float32` can be used.

#### Key Point
shape = layout, size = total number of elements, ndim = number of axes.

## Topic 8 – Data Types

### Step 1 – Understand the Concept
Every NumPy array has one data type (`dtype`) for all its elements. Common types are `int64`, `float64`, `float32`, `bool` and string types like `<U3`.

The type can be chosen while creating the array using `dtype=`, or changed later using `astype()`, which creates a new array. If a list has mixed types, NumPy converts all the values to one common type. For example, integers and floats together become floats.

### Step 2 – Demonstrate the Concept

#### Business Example
In a sales record, the price of a product needs decimals (`float`), such as 99.99, while the number of items sold is a whole number (`int`), such as 5. So the two columns need different data types.

### Step 3 – Implement the Concept

```python
ints = np.array([1, 2, 3])
floats = np.array([1.5, 2.5, 3.5])
mixed = np.array([1, 2.5, 3])           # int + float -> float
text = np.array(["a", "bb", "ccc"])
flags = np.array([True, False, True])

print("ints   :", ints.dtype)
print("floats :", floats.dtype)
print("mixed  :", mixed, "->", mixed.dtype)
print("text   :", text.dtype)
print("flags  :", flags.dtype)

# Choosing the type while creating
small = np.array([1, 2, 3], dtype=np.float32)
print("\nfloat32 array:", small, small.dtype, "| nbytes:", small.nbytes)

# Changing the type with astype
prices = np.array([99.99, 149.5, 20.75])
print("\nOriginal :", prices)
print("As int   :", prices.astype(int))
```

#### What does this program do?
The program shows the data type of integer, float, mixed, text and boolean arrays. It then creates a `float32` array and converts a price array to integers using `astype(int)`.

The right data type is important because it saves memory and avoids wrong calculations. In the output, the mixed array becomes `[1. 2.5 3.]` with type `float64`, and `astype(int)` gives `[99 149 20]`. It cuts off the decimals and does not round.

#### AI/ML Example
Deep learning models usually use `float32` instead of `float64`. This uses half the memory and trains faster. Text-coded numbers also have to be converted to numeric types before training.

#### Advantages
- Choosing a smaller type saves memory on large data.
- `astype()` makes conversion easy.

#### Limitations
- One array can only have one type.
- Converting a type can lose information, for example float to int removes the decimals.

#### Key Point
dtype = the type of all elements in an array. Use `astype()` to change it.

## Topic 9 – Indexing

### Step 1 – Understand the Concept
Indexing means getting an element from an array using its position.

- Indexing starts from **0**.
- Negative indexes count from the end, so `-1` is the last element.
- In a 2D array, `array[row, column]` is used.
- Boolean indexing selects the elements that satisfy a condition.
- Fancy indexing selects elements using a list of positions.

### Step 2 – Demonstrate the Concept

#### Business Example
From the revenue table in Topic 3, we want the revenue of Product 2 in Quarter 3. Product 2 is row index 1 and Quarter 3 is column index 2, so we use `revenue[1, 2]`, which is 21.

### Step 3 – Implement the Concept

```python
arr = np.array([10, 20, 30, 40, 50])
print("First element :", arr[0])
print("Third element :", arr[2])
print("Last element  :", arr[-1])

# 2D indexing: array[row, column]
print("\nRevenue table:\n", revenue)
print("Product 2, Quarter 3  :", revenue[1, 2])
print("Last row, last column :", revenue[-1, -1])

# 3D indexing: array[block, row, column]
print("\nTensor element [1, 0, 1]:", tensor[1, 0, 1])

# Boolean indexing
print("\nRevenue values greater than 15:", revenue[revenue > 15])

# Fancy indexing
print("Elements at positions 0, 2, 4 :", arr[[0, 2, 4]])

# ML-style: one sample and one feature from the customers data
print("\nThird customer (a whole row):", customers[2])
print("Income of third customer    :", customers[2, 1])
```

#### What does this program do?
The program gets elements by position from 1D, 2D and 3D arrays. It then uses a condition (boolean indexing) and a list of positions (fancy indexing) to select elements, and finally reads one customer's row and one feature from the `customers` data.

Indexing is used to take out exactly the data we need. The output shows `10, 30, 50`, then `21` and `9` from the revenue table, `6` from the tensor, the values above 15 as `[16 20 22 21 25]`, and `[47 80 12]` and `80` for the third customer.

#### AI/ML Example
In ML, indexing is used to pick one sample, one feature or one pixel from a dataset. Boolean indexing is used to filter data, for example "select all customers older than 40".

#### Advantages
- Fast access to any element.
- Boolean indexing makes filtering simple without loops.

#### Limitations
- An index outside the array gives an `IndexError`.
- Boolean and fancy indexing create copies, which use extra memory.

#### Key Point
Indexing starts at 0. Use `array[row, column]` for 2D arrays.

## Topic 10 – Slicing

### Step 1 – Understand the Concept
Slicing takes a part of an array using `start:stop:step`.

- `start` is included and `stop` is **not included**.
- If `start` is missing, it starts from the beginning. If `stop` is missing, it goes to the end.
- For 2D arrays, `array[row_slice, column_slice]` is used. A single `:` means "all" on that axis.

### Step 2 – Demonstrate the Concept

#### Numerical Example
Take the array `[10, 20, 30, 40, 50, 60, 70]`.

- `arr[1:4]` gives index 1, 2 and 3, so the result is `[20, 30, 40]`.
- `arr[::2]` takes every second element, so the result is `[10, 30, 50, 70]`.
- `arr[::-1]` reverses the array.

### Step 3 – Implement the Concept

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70])

print("arr[1:4]  :", arr[1:4])
print("arr[:3]   :", arr[:3])
print("arr[4:]   :", arr[4:])
print("arr[::2]  :", arr[::2])
print("arr[::-1] :", arr[::-1])

# 2D slicing
print("\nFirst two rows of revenue:\n", revenue[:2])
print("All rows, columns 1 to 2 :\n", revenue[:, 1:3])
print("Second column            :", revenue[:, 1])

# ML-style: split a small dataset into features X and target y
data = np.array([[5.1, 3.5, 0],
                 [4.9, 3.0, 0],
                 [6.2, 3.4, 1],
                 [5.9, 3.0, 1]])
X = data[:, :-1]     # all columns except the last
y = data[:, -1]      # only the last column
print("\nFeatures X:\n", X)
print("Target y   :", y)
```

#### What does this program do?
The program slices a 1D array in different ways, then slices rows and columns of a 2D array. At the end it splits a small dataset into features `X` (all columns except the last) and target `y` (the last column).

Slicing is used because it gets parts of the data quickly without loops. The output shows `[20 30 40]` for `arr[1:4]`, the reversed array for `arr[::-1]`, `X` with 4 rows and 2 columns, and `y` as `[0. 0. 1. 1.]`.

#### AI/ML Example
Separating features from the target (`X = data[:, :-1]`, `y = data[:, -1]`) is done in almost every ML project. Slicing is also used for splitting data into training and testing parts.

#### Advantages
- Very fast, because a slice does not copy the data.
- Short and readable code.

#### Limitations
- A slice is a view, so changing it also changes the original array (see Topic 13).

#### Key Point
Slicing = `start:stop:step`. The stop value is not included.

## Topic 11 – Reshaping

### Step 1 – Understand the Concept
Reshaping changes the shape of an array without changing its data. The only rule is that the total number of elements must stay the same.

If `-1` is used for one dimension, NumPy calculates that value automatically.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Suppose we have 12 chocolates. They can be arranged as 3 rows × 4 columns, 2 rows × 6 columns, or 4 rows × 3 columns. The chocolates are the same, only the arrangement changes.

### Step 3 – Implement the Concept

```python
arr = np.arange(1, 13)           # 12 elements
print("Original :", arr, "| shape:", arr.shape)

print("\nReshape to (3, 4):\n", arr.reshape(3, 4))
print("\nReshape to (2, 6):\n", arr.reshape(2, 6))
print("\nReshape to (4, -1):\n", arr.reshape(4, -1))

# ML-style: a single feature as a column
ages = np.array([22, 35, 58, 41])
print("\nAges shape before:", ages.shape, "| after reshape(-1, 1):", ages.reshape(-1, 1).shape)

# ML-style: a 28 x 28 image as one row of 784 values
image = np.zeros((28, 28))
print("Image shape after reshape(1, -1):", image.reshape(1, -1).shape)

# What happens when the sizes do not match
try:
    arr.reshape(5, 3)
except ValueError as error:
    print("\nError:", error)
```

#### What does this program do?
The program reshapes 12 numbers into different grids. It uses `-1` to let NumPy find one dimension, turns a 1D list into a column, turns a 28×28 image into one row, and finally shows the error when the new shape does not fit.

Reshaping is used because libraries expect data in specific shapes. The output shows 3×4 and 2×6 grids, a 4×3 grid for `(4, -1)`, ages going from `(4,)` to `(4, 1)`, the image shape `(1, 784)`, and a `ValueError` because 12 elements cannot become a 5×3 array.

#### AI/ML Example
Scikit-learn expects a single feature as a column, so `reshape(-1, 1)` is used often. Images are reshaped into rows of pixel values before being given to some models.

#### Advantages
- Changes the layout of data without changing the values.
- Usually fast, because it normally returns a view.

#### Limitations
- The total number of elements must stay the same.

#### Key Point
reshape() changes the shape, not the data. `-1` lets NumPy calculate that dimension.

## Topic 12 – Flatten

### Step 1 – Understand the Concept
Flattening converts a multi-dimensional array into a 1D array. NumPy has two common ways:

- `flatten()` always returns a **copy**.
- `ravel()` returns a **view** when possible. It is faster and uses no extra memory, but changing it can change the original array.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Think of a chocolate box arranged in rows and columns. Flattening is like taking all the chocolates out and placing them in one straight line, row after row.

### Step 3 – Implement the Concept

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

flat_copy = matrix.flatten()
flat_view = matrix.ravel()
print("flatten():", flat_copy)
print("ravel()  :", flat_view)

flat_copy[0] = 100
print("\nAfter changing the flatten() result, original matrix:\n", matrix)

flat_view[0] = 999
print("\nAfter changing the ravel() result, original matrix:\n", matrix)
```

#### What does this program do?
The program flattens a 2×3 matrix using both `flatten()` and `ravel()`. It then changes the first value of each result to see if the original matrix is affected.

This is done to show the difference between the two. Both give `[1 2 3 4 5 6]`. Changing the `flatten()` result does not affect the original, but changing the `ravel()` result changes the original matrix's first value to `999`.

#### AI/ML Example
Before an image is given to a fully-connected layer of a neural network, its pixels are flattened into a single 1D array.

#### Advantages
- Turns any array into 1D in one step.
- `flatten()` is safe and `ravel()` saves memory.

#### Limitations
- The 2D or 3D structure is lost after flattening.
- `ravel()` can change the original data by mistake.

#### Key Point
Use `flatten()` when you want a safe copy and `ravel()` when you only need to read the data.

## Topic 13 – Copy vs View

### Step 1 – Understand the Concept
- A **copy** is a new array with its own data. Changes to the copy do not change the original.
- A **view** is another way of looking at the same data. Changes made through the view also change the original.

Slicing returns a view. `.copy()`, `flatten()`, `astype()` and boolean/fancy indexing return copies. `np.shares_memory()` can be used to check if two arrays share data.

### Step 2 – Demonstrate the Concept

#### Real-World Example
A view is like a shared Google Doc: when one person edits it, everyone sees the change. A copy is like downloading the document: the edits on your downloaded file do not change the original.

### Step 3 – Implement the Concept

```python
original = np.array([1, 2, 3, 4, 5])

view = original[1:4]          # slice -> view
copy = original[1:4].copy()   # explicit copy

view[0] = 99                  # change through the view
copy[1] = 77                  # change the copy

print("original after changes:", original)
print("view                  :", view)
print("copy                  :", copy)

print("\nview shares memory with original?", np.shares_memory(original, view))
print("copy shares memory with original?", np.shares_memory(original, copy))
```

#### What does this program do?
The program creates a view (a slice) and a copy (using `.copy()`) from the same part of an array. It changes one value in each and then checks if the original array was affected.

The output shows that the original became `[ 1 99  3  4  5]` because of the view, while the copy did not change the original. `np.shares_memory` gives `True` for the view and `False` for the copy.

#### AI/ML Example
Suppose a part of the training data is sliced for preprocessing and then modified. Because the slice is a view, the original training data also changes silently, which can give wrong model results. Using `.copy()` avoids this.

#### Advantages
- Views save memory and time because no data is copied.
- Copies keep the original data safe.

#### Limitations
- Views can change the original data by mistake.
- Copies use extra memory for large arrays.

#### Key Point
View = same data, shared changes. Copy = separate data. Use `.copy()` when you want to modify safely.

## Topic 14 – Broadcasting

### Step 1 – Understand the Concept
Broadcasting lets NumPy do calculations between arrays of different shapes. NumPy automatically stretches the smaller array so that both shapes match, without actually copying the data.

NumPy compares the shapes from the last dimension to the first. Two dimensions are compatible if they are equal or if one of them is 1. If not, NumPy gives a `ValueError`.

### Step 2 – Demonstrate the Concept

#### Business Example
A shop has to add 18% GST to the prices of all products: 100, 200 and 300. The single number 1.18 is "stretched" and multiplied with every price, so there is no need to write it for each product.

### Step 3 – Implement the Concept

```python
# Array with a single number
prices = np.array([100, 200, 300])
print("Prices with 18% GST:", prices * 1.18)

# 2D array with a 1D row (the row is applied to every row)
marks = np.array([[70, 80, 90],
                  [60, 65, 75]])
bonus_per_subject = np.array([5, 0, 10])
print("\nMarks + bonus per subject:\n", marks + bonus_per_subject)

# 2D array with a column (the column is applied across every row)
bonus_per_student = np.array([[2], [4]])
print("\nMarks + bonus per student:\n", marks + bonus_per_student)

# ML-style: centering features by subtracting the column means
X = np.array([[170, 65],
              [180, 80],
              [160, 55]])      # height (cm), weight (kg)
col_means = X.mean(axis=0)
print("\nColumn means:", col_means)
print("Centered data:\n", X - col_means)

# Shapes that cannot be broadcast
try:
    np.ones((3, 2)) + np.ones((3,))
except ValueError as error:
    print("\nBroadcasting error:", error)
```

#### What does this program do?
The program uses broadcasting in four ways: multiplying an array by one number, adding a row to every row of a 2D array, adding a column across every row, and subtracting the column means from a dataset. The last part shows two shapes that cannot be broadcast.

Broadcasting is used to avoid loops and to avoid making large repeated arrays. The output shows prices of `[118. 236. 354.]`, the bonus added to the marks, and the centered data. The last example gives a `ValueError` because the last dimensions 2 and 3 are not equal and neither is 1.

#### AI/ML Example
Feature scaling is a very common use. Subtracting the mean of each column from the dataset (`X - X.mean(axis=0)`) works because of broadcasting. It is also used to add a bias to the output of a neural network layer.

#### Advantages
- No loops are needed, so the code is short and fast.
- Does not create large repeated copies of data.

#### Limitations
- Shapes must follow the broadcasting rules.
- It can hide mistakes when the shapes happen to match in an unexpected way.

#### Key Point
Broadcasting compares shapes from the last dimension. Each pair must be equal or one must be 1.

## Topic 15 – Array Operations

### Step 1 – Understand the Concept
NumPy operations work element by element on whole arrays.

- **Arithmetic:** `+`, `-`, `*`, `/`, `//`, `%`, `**`
- **Comparison:** `>`, `<`, `==`, `!=` (gives a boolean array)
- **Logical:** `&` (and), `|` (or), with brackets around each condition
- **Aggregation:** `sum()`, `min()`, `max()`, `mean()`. With `axis=0` the calculation is done down each column, and with `axis=1` across each row.

The `*` operator is element-wise multiplication. It is not matrix multiplication, which comes in Notebook 2.

### Step 2 – Demonstrate the Concept

#### Business Example
A shop has the prices 50, 120, 80, 200 and the quantities sold 10, 3, 7, 2. The revenue of every product is price × quantity, which NumPy calculates for all products in one step: `500, 360, 560, 400`.

### Step 3 – Implement the Concept

```python
price = np.array([50, 120, 80, 200])
qty = np.array([10, 3, 7, 2])

# Arithmetic
revenue_each = price * qty
print("Revenue per product:", revenue_each)
print("Price + 10         :", price + 10)

# Comparison and filtering
print("\nRevenue > 400?                        :", revenue_each > 400)
print("Prices of products with revenue > 400 :", price[revenue_each > 400])

# Logical
mask = (price > 60) & (qty > 2)
print("\nPrice > 60 AND qty > 2:", mask)

# Aggregation with axis
scores = np.array([[80, 90, 70],
                   [60, 75, 85]])
print("\nTotal of all values    :", scores.sum())
print("Column totals (axis=0) :", scores.sum(axis=0))
print("Row totals    (axis=1) :", scores.sum(axis=1))

# ML-style: prediction error
y_true = np.array([3.0, 5.0, 2.5, 7.0])
y_pred = np.array([2.5, 5.0, 4.0, 8.0])
error = y_pred - y_true
print("\nPrediction error   :", error)
print("Mean absolute error:", np.abs(error).mean())
```

#### What does this program do?
The program calculates revenue from price and quantity, compares and filters values, combines two conditions, and uses `sum()` along different axes. At the end it calculates the error and Mean Absolute Error of some predictions.

These operations replace loops and are used daily in data work. The output shows revenue `[500 360 560 400]`, the comparison `[True False True False]`, column totals `[140 165 155]`, row totals `[240 220]`, errors `[-0.5 0. 1.5 1.]` and a Mean Absolute Error of `0.75`.

#### AI/ML Example
When a model makes predictions, `y_pred - y_true` calculates the error of every sample at once. Loss functions like Mean Absolute Error and Mean Squared Error are built from these array operations.

#### Advantages
- Works on whole arrays without loops.
- Short, readable code.

#### Limitations
- Both arrays must have compatible shapes.
- Using `and` / `or` instead of `&` / `|` on arrays gives an error.

#### Key Point
NumPy operations are element-wise. `axis=0` works on columns and `axis=1` works on rows.

## Topic 16 – Universal Functions (ufuncs)

### Step 1 – Understand the Concept
A universal function (ufunc) is a NumPy function that works element by element on an array. It is written in fast compiled code and supports broadcasting.

Examples are `np.add`, `np.sqrt`, `np.square`, `np.exp`, `np.log`, `np.abs` and `np.maximum`. Operators like `+` and `*` are also ufuncs underneath. Ufuncs also have extra methods like `reduce` (combine all values into one) and `accumulate` (running result at every step).

### Step 2 – Demonstrate the Concept

#### Numerical Example
For the array `[1, 4, 9, 16]`:

- `np.sqrt` gives `[1, 2, 3, 4]`.
- `np.add.reduce` adds all values and gives `30`.
- `np.add.accumulate` gives the running total `[1, 5, 14, 30]`.

### Step 3 – Implement the Concept

```python
nums = np.array([1, 4, 9, 16])

print("sqrt    :", np.sqrt(nums))
print("square  :", np.square(nums))
print("log     :", np.log(nums))
print("add     :", np.add(nums, 10))
print("maximum :", np.maximum(nums, 5))

print("\nreduce     :", np.add.reduce(nums))
print("accumulate :", np.add.accumulate(nums))

# ML-style: sigmoid and ReLU activation functions
x = np.array([-2.0, -1.0, 0.0, 1.0, 2.0])
sigmoid = 1 / (1 + np.exp(-x))
relu = np.maximum(0, x)
print("\nInput   :", x)
print("Sigmoid :", np.round(sigmoid, 3))
print("ReLU    :", relu)
```

#### What does this program do?
The program applies several ufuncs to an array, uses `reduce` and `accumulate`, and then calculates the sigmoid and ReLU activation functions for a small input array.

Ufuncs are used because they apply a calculation to every value quickly, without loops. The output shows `[1. 2. 3. 4.]` for the square root, `30` for `reduce`, `[1 5 14 30]` for `accumulate`, sigmoid values between 0 and 1 (exactly `0.5` for input 0), and ReLU turning all negative values into 0.

#### AI/ML Example
Activation functions in neural networks are built using ufuncs: sigmoid uses `np.exp` and ReLU uses `np.maximum`. `np.log` is used in loss functions like log-loss, and `np.log` / `np.sqrt` are used to transform skewed data.

#### Advantages
- Fast and vectorized.
- Support broadcasting.
- Available for most common maths operations.

#### Limitations
- They work element by element, so operations that need to compare different rows of data need other tools.
- Invalid inputs give `nan` or warnings, for example `np.log(0)` or `np.sqrt(-1)`.

#### Key Point
A ufunc applies a fast operation to every element of an array. `reduce` and `accumulate` give single and running results.

## Best Practices
- Always import NumPy as `import numpy as np`.
- Check `shape`, `ndim` and `dtype` right after creating or loading an array.
- Use vectorized operations and ufuncs instead of Python `for` loops.
- Use `.copy()` when you want to change a slice without changing the original.
- Choose a suitable `dtype` (such as `float32`) to save memory on large data.
- Use `reshape(-1, 1)` or `reshape(1, -1)` instead of typing sizes by hand.

## Advantages of NumPy
- Very fast because of compiled code and vectorization.
- Uses less memory than Python lists.
- Has many mathematical and statistical functions.
- Supports multi-dimensional data and broadcasting.
- Is the base of Pandas, Scikit-learn, TensorFlow and PyTorch.

## Limitations of NumPy
- All elements must be of the same data type.
- The size of an array is fixed after creation.
- Not made for labelled or mixed-type tables (Pandas is better).
- Works on the CPU by default. GPU libraries such as PyTorch are used for GPU work.
- The whole array has to fit in memory.

## Interview Questions

**1. What is the difference between a Python list and a NumPy array?**</br>
A list can store different types of values and is slower for numerical work. A NumPy array stores values of one type together in memory, uses less memory and supports fast vectorized operations.

**2. What is the difference between `shape`, `size` and `ndim`?**</br>
`shape` is the size along each axis, `size` is the total number of elements and `ndim` is the number of axes. For a 3×4 array, the shape is `(3, 4)`, the size is 12 and `ndim` is 2.

**3. What is the difference between a copy and a view?**</br>
A copy has its own data, so changing it does not affect the original. A view shares the same data, so changing it also changes the original. Slicing gives a view and `.copy()` gives a copy.

**4. What is broadcasting? Explain with an example.**</br>
Broadcasting lets NumPy do calculations on arrays with different shapes by stretching the smaller array. Shapes are compared from the last dimension, and each pair must be equal or one must be 1. For example, adding `[10, 20, 30]` to a 2×3 array adds it to every row.

**5. What is the difference between `flatten()` and `ravel()`?**</br>
`flatten()` always returns a copy. `ravel()` returns a view when possible, so it is faster, but changing it can change the original array.

**6. What does `reshape(-1, 1)` do?**</br>
It changes the array into one column with as many rows as needed. The `-1` tells NumPy to calculate that dimension automatically. It is often used to prepare a single feature for Scikit-learn.

**7. What is the difference between `arange` and `linspace`?**</br>
`arange(start, stop, step)` uses a step size and does not include the stop value. `linspace(start, stop, num)` gives a fixed number of equally spaced values and includes the stop value.

**8. What is a ufunc?**</br>
A ufunc (universal function) is a NumPy function such as `np.sqrt`, `np.exp` or `np.add` that works element by element on arrays. It is written in compiled code, so it is fast.

**9. Why is NumPy faster than Python lists?**</br>
NumPy runs in compiled C code, stores values of the same type next to each other in memory, and uses vectorization instead of Python loops.
