## Topic 1 – What is Pandas?

### Step 1 – Understand the Concept
Pandas is an open-source Python library used to work with **tabular data**, which means data in rows and columns, like an Excel sheet or a database table. The name comes from "Panel Data".

Pandas is built on top of NumPy. Its two main objects are:

- **Series** – one column of data with labels.
- **DataFrame** – a full table with rows and columns, which can be thought of as a group of Series.

### Step 2 – Demonstrate the Concept

#### Business Example
A company keeps the details of its employees: name, age, department and salary. In Excel this is a sheet with 4 columns. In Pandas, the same sheet becomes a DataFrame, and the company can find the average salary or the highest salary with one line of code.

### Step 3 – Implement the Concept

```python
employees = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Sana"],
    "Age": [25, 32, 29, 41, 36],
    "Department": ["HR", "IT", "IT", "Finance", "HR"],
    "Salary": [40000, 65000, 58000, 72000, 45000]
})

print(employees)
print("\nType  :", type(employees))
print("Shape :", employees.shape)
print("\nAverage salary:", employees["Salary"].mean())
print("Highest salary:", employees["Salary"].max())
```

#### What does this program do?
The program creates a table of 5 employees using `pd.DataFrame()` and prints it. It then prints the type and shape of the table, and finds the average and highest salary by using the column name `"Salary"`.

Pandas is used because the data stays in a clear table with column names, and calculations are done with the column name instead of position numbers. The output shows a table with 5 rows and 4 columns, the shape `(5, 4)`, an average salary of `56000.0` and a highest salary of `72000`.

#### AI/ML Example
Almost every ML project starts by loading a dataset (for example a CSV file of customers or houses) into a Pandas DataFrame. The data is then explored, cleaned and prepared using Pandas before it is given to a model.

#### Advantages
- Handles tables with column names and different data types easily.
- Reads data from many sources like CSV, Excel and databases.
- Has many ready-made functions for cleaning and analysing data.

#### Limitations
- Works in memory, so very large datasets that do not fit in RAM are a problem.
- Slower than NumPy for pure numerical calculations on large arrays.

#### Key Point
Pandas = a Python library for working with tabular data. Its main objects are the Series (one column) and the DataFrame (a table).

## Topic 2 – Why Pandas?

### Step 1 – Understand the Concept
NumPy arrays are fast, but they have problems with real-world data:

- All elements must have the **same data type**, so text and numbers together are all turned into text.
- There are **no column names**, so we have to remember that "column 2 is salary".

Pandas solves these problems:

- Each column can have its **own data type**.
- Columns and rows have **labels** (names).
- Easy handling of **missing values**, filtering, grouping and merging.
- Easy reading and writing of files.

### Step 2 – Demonstrate the Concept

#### Real-World Example
Think of a school register. It has columns like Name, Age and Marks, and every column has a heading. If we removed the headings and wrote everything as plain text, it would be very hard to find the average marks. A DataFrame is like the register with headings, and a NumPy array is like the register without them.

### Step 3 – Implement the Concept

```python
rows = [["Asha", 25, 40000],
        ["Ravi", 32, 65000],
        ["Meena", 29, 58000]]

# With NumPy: text and numbers are mixed, so everything becomes text
arr = np.array(rows)
print(arr)
print("NumPy data type:", arr.dtype)

try:
    arr[:, 1].mean()
except Exception as error:
    print("Mean of the Age column gives an error:", type(error).__name__)

# With Pandas: each column keeps its own type and has a name
df = pd.DataFrame(rows, columns=["Name", "Age", "Salary"])
print("\n", df)
print("\nColumn data types:\n", df.dtypes)
print("\nAverage salary:", df["Salary"].mean())
print("\nPeople older than 26:\n", df[df["Age"] > 26])
```

#### What does this program do?
The program puts the same data (names, ages, salaries) into a NumPy array and into a Pandas DataFrame. With NumPy, every value becomes text, so calculating the mean of the Age column fails. With Pandas, every column keeps its own type, and we can use column names to calculate an average and to filter rows.

This is done to show why Pandas is better for tables with mixed data. The output shows the NumPy data type `<U21` (text), a `TypeError` for the mean, a DataFrame with `Age` and `Salary` as `int64`, an average salary of `54333.33`, and a filtered table with only Ravi and Meena (the people older than 26).

#### AI/ML Example
Real ML datasets have mixed data: numbers (age, income), categories (city, gender) and dates. Pandas is used for Exploratory Data Analysis (EDA), data cleaning and feature engineering, and the cleaned data is then converted to NumPy arrays for the model.

#### Advantages
- Column names make code easy to read.
- Different data types in one table.
- Built-in support for missing values and filtering.

#### Limitations
- Uses more memory than NumPy arrays.
- Slower than NumPy for heavy number crunching.

#### Key Point
Use NumPy for pure numerical arrays and Pandas for labelled tables with mixed data types.

## Topic 3 – Series

### Step 1 – Understand the Concept
A Series is a one-dimensional labelled array. It is like one column of a table. It has:

- **values** – the data (stored as a NumPy array),
- **index** – a label for each value (0, 1, 2 ... by default, or custom labels),
- **dtype** – the data type of the values,
- an optional **name**.

A Series can be created from a list, a dictionary or a NumPy array using `pd.Series()`. Calculations on a Series work on all the values at once, like in NumPy.

### Step 2 – Demonstrate the Concept

#### Real-World Example
The marks of five students: Asha 78, Ravi 85, Meena 62, Kiran 90 and Sana 71. Here the student names are the labels (index) and the marks are the values, so the marks of any student can be found by name.

### Step 3 – Implement the Concept

```python
# Series with the default index (0, 1, 2 ...)
marks = pd.Series([78, 85, 62, 90, 71])
print(marks)
print("dtype:", marks.dtype)

# Series with custom labels and a name
named = pd.Series([78, 85, 62, 90, 71],
                  index=["Asha", "Ravi", "Meena", "Kiran", "Sana"],
                  name="Marks")
print("\n", named)
print("\nMarks of Ravi (by label)  :", named["Ravi"])
print("First value (by position) :", named.iloc[0])

# Series from a dictionary: keys become the labels
daily = pd.Series({"Mon": 120, "Tue": 135, "Wed": 150})
print("\n", daily)

# Calculations work on the whole Series
print("\nMean marks      :", named.mean())
print("Marks + 5 bonus :\n", named + 5)
print("Marks above 75  :\n", named[named > 75])

# ML-style: a single column of a DataFrame is a Series
age_column = employees["Age"]
print("\nType of employees['Age']:", type(age_column))
print("Index of named Series   :", named.index.tolist())
print("Values as NumPy array   :", type(named.values))
```

#### What does this program do?
The program creates a Series with the default index and another one with student names as the index. It shows how to get a value by label (`named["Ravi"]`) and by position (`named.iloc[0]`), creates a Series from a dictionary, and does calculations such as the mean, adding 5 marks and filtering marks above 75. At the end it shows that a column of a DataFrame is a Series, and that the values inside a Series are a NumPy array.

Labels are used so that data can be found by name and not by position. The output shows the mean `77.2`, all marks increased by 5 (`83, 90, 67, 95, 76`), and only Asha, Ravi and Kiran in the filtered result. The type of `employees['Age']` is a Series, and `named.values` is a NumPy array. (`loc` and `iloc` are explained fully in Notebook 7.)

#### AI/ML Example
In an ML dataset, every feature (like age or salary) is a column, which is a Series. The target variable (like "purchased: yes/no") is also a Series. Many calculations in data preparation are done on one Series at a time.

#### Advantages
- Values can be found by meaningful labels.
- Fast calculations on all values at once.
- Easy to convert to NumPy using `.values` or `.to_numpy()`.

#### Limitations
- It has only one dimension (one column). For many columns a DataFrame is needed.
- All values in a Series share one data type.

#### Key Point
A Series is one labelled column of data: values + index. A column of a DataFrame is a Series.

## Topic 4 – DataFrame

### Step 1 – Understand the Concept
A DataFrame is a two-dimensional table with rows and columns. Each column is a Series, and different columns can have different data types. It has:

- an **index** (row labels),
- **columns** (column names),
- **values** (the data).

Useful attributes are `shape` (rows, columns), `columns`, `index`, `dtypes` and the method `to_numpy()`, which converts the table to a NumPy array.

### Step 2 – Demonstrate the Concept

#### Business Example
A shop keeps a table of its products with the columns Product, Price and Quantity. Each row is one product and each column is one detail. This table is a DataFrame.

### Step 3 – Implement the Concept

```python
sales = pd.DataFrame({
    "Product": ["Pen", "Book", "Bag", "Bottle"],
    "Price": [10, 250, 600, 120],
    "Quantity": [50, 12, 5, 20]
})

print(sales)
print("\nShape:", sales.shape, "-> (rows, columns)")

# One column = Series, a list of columns = DataFrame
print("\nsales['Price'] is a       :", type(sales["Price"]))
print("sales[['Product','Price']] is a:", type(sales[["Product", "Price"]]))

# Converting to NumPy
print("\nWhole table to NumPy (mixed types):\n", sales.to_numpy())
numbers = sales[["Price", "Quantity"]].to_numpy()
print("\nOnly number columns to NumPy:\n", numbers)
print("Data type:", numbers.dtype)

# ML-style: features X and target y
study = pd.DataFrame({"Hours": [1, 2, 3, 4, 5],
                      "Marks": [45, 50, 60, 65, 75]})
X = study[["Hours"]]       # features: a DataFrame (double brackets)
y = study["Marks"]         # target: a Series (single brackets)
print("\nX is a", type(X).__name__, "with shape", X.shape)
print("y is a", type(y).__name__, "with shape", y.shape)
```

#### What does this program do?
The program creates a product table and prints its shape. It shows that selecting one column with single brackets gives a Series, while selecting with a list in double brackets gives a DataFrame. It converts the table to a NumPy array with `to_numpy()`, first with all columns and then with only the number columns. At the end it separates a small study dataset into features `X` and target `y`.

`to_numpy()` is used because many libraries need NumPy arrays. The output shows the shape `(4, 3)`, a Series for `sales['Price']` and a DataFrame for the list of columns. The whole table becomes an array of mixed objects, while the number columns become an `int64` array of shape `(4, 2)`. `X` is a DataFrame with shape `(5, 1)` and `y` is a Series with shape `(5,)`.

#### AI/ML Example
In ML, a DataFrame is split into the feature columns `X` (inputs) and the target column `y` (what we want to predict). Models in Scikit-learn accept DataFrames and Series, and convert them to NumPy arrays internally.

#### Advantages
- Holds a full dataset with column names and mixed types.
- Easy to select, filter and calculate by column name.
- Easy conversion to and from NumPy.

#### Limitations
- Uses more memory than a NumPy array of the same data.
- Needs the whole table in memory.

#### Key Point
A DataFrame is a table of rows and columns where every column is a Series. Single brackets `df["col"]` give a Series and double brackets `df[["col"]]` give a DataFrame.

## Topic 5 – Creating DataFrames

### Step 1 – Understand the Concept
A DataFrame can be created from many kinds of data using `pd.DataFrame()`:

| Source | How |
|---|---|
| Dictionary of lists | keys become column names, lists become columns |
| List of lists | give the column names with `columns=[...]` |
| List of dictionaries | each dictionary is one row (missing keys become `NaN`) |
| NumPy array | give the column names with `columns=[...]` |
| Dictionary of Series | each Series becomes a column and the index is matched |
| Empty DataFrame | give only the `columns=[...]` |

The row labels can be chosen with the `index=` argument. DataFrames can also be created by reading files, which is explained in the Reading Data topic.

### Step 2 – Demonstrate the Concept

#### Business Example
An online shop receives orders as records, and each record is a dictionary with the order number, item and quantity. One order does not have a quantity. Putting the list of records into a DataFrame gives a table, and the missing value is shown as `NaN` (Not a Number).

### Step 3 – Implement the Concept

```python
# 1. Dictionary of lists
d1 = pd.DataFrame({"Name": ["Asha", "Ravi"], "Age": [25, 32]})
print("From a dictionary of lists:\n", d1)

# 2. List of lists with column names
d2 = pd.DataFrame([["Pen", 10], ["Book", 250]], columns=["Item", "Price"])
print("\nFrom a list of lists:\n", d2)

# 3. List of dictionaries (business example: order records)
orders = [{"OrderID": 1, "Item": "Pen", "Qty": 10},
          {"OrderID": 2, "Item": "Book", "Qty": 2},
          {"OrderID": 3, "Item": "Bag"}]
print("\nFrom a list of dictionaries:\n", pd.DataFrame(orders))

# 4. NumPy array (ML-style: random features)
np.random.seed(42)
data = np.random.randint(1, 100, (4, 3))
features = pd.DataFrame(data, columns=["feature_1", "feature_2", "feature_3"])
print("\nFrom a NumPy array:\n", features)

# 5. Dictionary of Series
names = pd.Series(["Asha", "Ravi"])
ages = pd.Series([25, 32])
print("\nFrom Series:\n", pd.DataFrame({"Name": names, "Age": ages}))

# 6. Custom row labels and an empty DataFrame
print("\nWith custom index:\n", pd.DataFrame({"Marks": [78, 85]}, index=["Asha", "Ravi"]))
empty = pd.DataFrame(columns=["A", "B"])
print("\nEmpty DataFrame shape:", empty.shape)
```

#### What does this program do?
The program creates DataFrames in six ways: from a dictionary of lists, a list of lists, a list of dictionaries, a NumPy array, Series, and with custom row labels. It also creates an empty DataFrame that has only column names.

Different sources are used because data comes in many forms in real work. The output shows the same kind of table each time. In the list of dictionaries, the third order has no quantity, so the value is `NaN` and the `Qty` column becomes decimal numbers (`10.0`, `2.0`, `NaN`). The NumPy array gives a table of random whole numbers with the names `feature_1` to `feature_3`, and the empty DataFrame has the shape `(0, 2)` (no rows, 2 columns).

#### AI/ML Example
After computations in NumPy (for example a random array or the output of a model), the array is often put into a DataFrame with column names so it is easier to read, inspect and save.

#### Advantages
- Many ways to create a table from the data we already have.
- Missing values in records are filled with `NaN` automatically.

#### Limitations
- Creating a DataFrame from a very large Python list can be slow and memory heavy.
- Lists in a dictionary must have the same length, otherwise Pandas gives a `ValueError`.

#### Key Point
`pd.DataFrame()` can build a table from dictionaries, lists, NumPy arrays and Series. Use `columns=` and `index=` to set the names.

## Topic 6 – Data Types

### Step 1 – Understand the Concept
Every column of a DataFrame has a data type (`dtype`). It can be checked with `df.dtypes`.

| Pandas dtype | Meaning | Example |
|---|---|---|
| `int64` | whole numbers | age, quantity |
| `float64` | decimal numbers (also used when `NaN` is present) | salary, price |
| `object` / `str` | text | names, cities |
| `bool` | True / False | is_active |
| `datetime64` | dates and times | joining date |

Types can be changed with `astype()`. If a column has text that should be a number, `pd.to_numeric()` is used, and with `errors="coerce"` the values that cannot be converted become `NaN` instead of giving an error.

### Step 2 – Demonstrate the Concept

#### Business Example
A company table has a name (text), age (whole number), salary (decimal), active status (True/False) and joining date (date). Each of these needs a different data type, and Pandas gives each column its own type.

### Step 3 – Implement the Concept

```python
people = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena"],
    "Age": [25, 32, 29],
    "Salary": [40000.5, 65000.0, 58000.25],
    "Active": [True, False, True],
    "Joined": pd.to_datetime(["2020-01-15", "2019-06-01", "2021-03-20"])
})
print(people.dtypes)

# Changing types with astype
print("\nAge as float :", people["Age"].astype(float).dtype)
print("Salary as int:", people["Salary"].astype(int).tolist(), "(decimals are cut off)")

# ML-style: numbers stored as text, with one wrong value
prices = pd.Series(["100", "250", "abc", "400"])
print("\nBefore:", prices.dtype)
clean = pd.to_numeric(prices, errors="coerce")
print("After to_numeric:\n", clean)
print("dtype:", clean.dtype)
```

#### What does this program do?
The program creates a table with text, whole numbers, decimals, booleans and dates, and prints the data type of every column. It then changes types with `astype()` and converts a Series of numbers stored as text into real numbers using `pd.to_numeric()` with `errors="coerce"`.

Checking and fixing types is important because calculations only work on the right types. The output shows `str` (or `object`), `int64`, `float64`, `bool` and `datetime64` for the five columns. `astype(float)` gives `float64`, and `astype(int)` cuts the decimals (`[40000, 65000, 58000]`). The text `"abc"` cannot be a number, so it becomes `NaN`, and the result is `[100.0, 250.0, NaN, 400.0]` with the type `float64`.

#### AI/ML Example
Data from files often has numbers stored as text (for example "1,200" or "N/A"). Before training a model, these columns must be converted to numeric types. `pd.to_numeric(errors="coerce")` is commonly used, and the `NaN` values created are handled later as missing data.

#### Advantages
- Each column can have its own type.
- The right types save memory and make calculations possible.

#### Limitations
- Wrong types (like numbers stored as text) cause errors or wrong results and need to be fixed manually.
- `astype()` can lose information, for example float to int cuts the decimals.

#### Key Point
Check `df.dtypes` after loading data. Use `astype()` to change types and `pd.to_numeric(errors="coerce")` to clean numbers stored as text.

## Topic 7 – Index

### Step 1 – Understand the Concept
The index is the set of labels for the rows of a Series or DataFrame. By default it is a `RangeIndex` of numbers 0, 1, 2 and so on, but it can be changed to something meaningful, like a name or an ID.

- `df.index` shows the index.
- `df.set_index("col")` makes a column the index.
- `df.reset_index()` moves the index back to a normal column and restores the numbers 0, 1, 2.
- `df.loc[label]` selects a row by its label.

Pandas also uses the index for **alignment**: when two Series are added, the values are matched by their labels and not by their positions.

### Step 2 – Demonstrate the Concept

#### Real-World Example
In a school register, every student has a roll number. Finding a student by roll number is easier than counting rows from the top. In the same way, using the employee name as the index lets us find an employee by name.

### Step 3 – Implement the Concept

```python
print("Default index:", employees.index)

# Making Name the index
emp_named = employees.set_index("Name")
print("\n", emp_named)
print("\nRow of Ravi:\n", emp_named.loc["Ravi"])
print("\nIndex labels:", emp_named.index.tolist())

# Going back to the default index
print("\nColumns after reset_index():", emp_named.reset_index().columns.tolist())

# ML-style: index alignment when adding two Series
a = pd.Series([10, 20, 30], index=["a", "b", "c"])
b = pd.Series([1, 2, 3], index=["b", "c", "d"])
print("\na + b:\n", a + b)
```

#### What does this program do?
The program shows the default index of the employees table. It then makes `Name` the index with `set_index`, selects the row of Ravi with `loc`, and uses `reset_index()` to turn the index back into a column. At the end it adds two Series that have different labels.

The index is used to find and match data by meaningful labels. The output shows `RangeIndex(start=0, stop=5, step=1)` as the default index, a table where the names are on the left, and Ravi's row (`Age 32`, `Department IT`, `Salary 65000`). After `reset_index()`, `Name` is a normal column again. In `a + b`, only `b` and `c` exist in both Series, so the result is `21.0` and `32.0`, while `a` and `d` become `NaN`.

#### AI/ML Example
In time series data, the date is used as the index so that data can be selected and aligned by time. When combining two datasets or results from different steps, Pandas uses the index to match the right rows. Wrong or mismatched indexes are a common reason for unexpected `NaN` values.

#### Advantages
- Fast lookup of rows by label.
- Automatic alignment of data when combining Series or DataFrames.

#### Limitations
- Labels can be duplicated, which can give unexpected results when selecting.
- After filtering or shuffling, the old index numbers stay, so `reset_index(drop=True)` may be needed.

#### Key Point
The index labels the rows. Pandas matches data by index labels, not by position. Use `set_index()` and `reset_index()` to change it.

## Topic 8 – Columns

### Step 1 – Understand the Concept
Columns are the named fields of a DataFrame. `df.columns` shows the names, and column names are used to select data:

- `df["col"]` – one column (gives a Series).
- `df[["col1", "col2"]]` – several columns (gives a DataFrame).
- `"col" in df.columns` – checks if a column exists.

Column names are case-sensitive and spaces count, so `"Salary"` and `"salary"` are different names. Selecting a name that does not exist gives a `KeyError`.

### Step 2 – Demonstrate the Concept

#### Business Example
A customer table has the columns Name, Age, Department and Salary. A manager who only wants to see names and salaries selects just those two columns, and ignores the rest.

### Step 3 – Implement the Concept

```python
print("Columns        :", employees.columns.tolist())
print("Number of columns:", len(employees.columns))
print("Is 'Salary' a column?", "Salary" in employees.columns)

print("\nOne column (Series):\n", employees["Name"])
print("\nTwo columns (DataFrame):\n", employees[["Name", "Salary"]])

# ML-style: feature column and target column
X = employees[["Age"]]
y = employees["Salary"]
print("\nFeatures X shape:", X.shape)
print("Target y shape  :", y.shape)

# Column names are case-sensitive
try:
    employees["salary"]
except KeyError as error:
    print("\nKeyError:", error)
```

#### What does this program do?
The program prints the column names, counts them and checks if a column exists. It selects one column and two columns, separates a feature `X` (Age) and a target `y` (Salary), and finally shows the `KeyError` that comes from writing a column name in the wrong case.

Column names are used because they are clearer and safer than position numbers. The output shows `['Name', 'Age', 'Department', 'Salary']`, `4` columns, `True` for `'Salary'`, a Series for `Name`, a two-column DataFrame, `X` with shape `(5, 1)`, `y` with shape `(5,)`, and `KeyError: 'salary'`.

#### AI/ML Example
In ML, specific columns are chosen as input features and one column as the target. Selecting columns by name makes the code easy to read and avoids mistakes such as accidentally including the target in the features.

#### Advantages
- Clear, readable code because columns are used by name.
- Easy to select any group of columns.

#### Limitations
- Spelling and case mistakes cause a `KeyError`.
- Extra spaces in column names from files (like `"Salary "`) can cause hard-to-find errors.

#### Key Point
Use `df["col"]` for one column (Series) and `df[["col1", "col2"]]` for several columns (DataFrame). Column names are case-sensitive.

## Topic 9 – Reading Data

### Step 1 – Understand the Concept
In real projects, data usually comes from files. The most common file is a CSV (Comma Separated Values) file, where every line is a row and values are separated by commas. `pd.read_csv("file.csv")` reads it into a DataFrame.

Useful arguments of `read_csv`:

| Argument | Use |
|---|---|
| `sep` | the separator, if it is not a comma (for example `sep=";"`) |
| `usecols` | read only the listed columns |
| `nrows` | read only the first n rows |
| `index_col` | use a column as the index |

If only the file name is given, the file must be in the same folder as the notebook, otherwise the full path is needed. All file types (Excel, JSON, Parquet and more) and file paths are covered in Notebook 5.

### Step 2 – Demonstrate the Concept

#### Business Example
The HR department keeps the employee list in a CSV file called `employees.csv`. Reading it with Pandas turns the file into a DataFrame, which can then be analysed.

### Step 3 – Implement the Concept
Since a real file is needed, I first create a small `employees.csv` file with plain Python. In real projects, the file already exists.

```python
csv_text = '''Name,Age,Department,Salary
Asha,25,HR,40000
Ravi,32,IT,65000
Meena,29,IT,58000
Kiran,41,Finance,72000
Sana,36,HR,45000
'''

with open("employees.csv", "w") as file:
    file.write(csv_text)

# Reading the whole file
df = pd.read_csv("employees.csv")
print(df)
print("\nData types:\n", df.dtypes)

# Reading only some columns, only some rows, and using a column as index
print("\nusecols:\n", pd.read_csv("employees.csv", usecols=["Name", "Salary"]))
print("\nnrows=2:\n", pd.read_csv("employees.csv", nrows=2))
print("\nindex_col='Name':\n", pd.read_csv("employees.csv", index_col="Name"))

# A file that uses a semicolon as the separator
semicolon_text = "Name;Age\nAsha;25\nRavi;32\n"
print("\nWithout sep:\n", pd.read_csv(io.StringIO(semicolon_text)))
print("\nWith sep=';':\n", pd.read_csv(io.StringIO(semicolon_text), sep=";"))
```

#### What does this program do?
The program first writes a small CSV file to the current folder. It then reads the file into a DataFrame with `pd.read_csv()` and prints the data types. After that it reads only two columns (`usecols`), only the first 2 rows (`nrows`), and uses `Name` as the index (`index_col`). At the end it reads text that uses `;` as the separator, first without `sep` and then with `sep=";"`.

Pandas finds the data types by itself while reading. ## Topic 10 – Writing Data

### Step 1 – Understand the Concept
`df.to_csv("file.csv")` saves a DataFrame to a CSV file. By default it also saves the index as the first column. This is usually not wanted, so `index=False` is used.

If a file saved with the index is read again, the saved index comes back as an extra column named `Unnamed: 0`. This can be avoided by using `index=False` when saving, or by using `index_col=0` when reading.

Saving to other formats (Excel, JSON, Parquet and more) is covered in Notebook 5.

### Step 2 – Demonstrate the Concept

#### Business Example
After analysing the employee data, the HR team wants to send the table to the management as a file. They save the DataFrame as a CSV file, which can be opened in Excel.

### Step 3 – Implement the Concept

```python
# Saving without the index (recommended)
employees.to_csv("employees_saved.csv", index=False)
print("File content with index=False:")
print(open("employees_saved.csv").read())
print("Shape after reading back:", pd.read_csv("employees_saved.csv").shape)

# Saving with the index (default)
employees.to_csv("employees_with_index.csv")
print("\nFile content with the index saved:")
print(open("employees_with_index.csv").read())

print("Columns read back  :", pd.read_csv("employees_with_index.csv").columns.tolist())
print("Fixed with index_col=0:", pd.read_csv("employees_with_index.csv", index_col=0).columns.tolist())
```

#### What does this program do?
The program saves the employees table to a CSV file with `index=False`, prints the content of the file, and reads it back to check the shape. It then saves the table again with the default settings (with the index), shows that the file has an extra first column, and shows the extra `Unnamed: 0` column that appears when reading it back, and how `index_col=0` removes it.

`index=False` is used so the file only contains the real data. The output shows a file with the header `Name,Age,Department,Salary` and 5 rows, and a shape of `(5, 4)` after reading it back. The file saved with the index has an extra first column with the numbers 0 to 4, the columns read back are `['Unnamed: 0', 'Name', 'Age', 'Department', 'Salary']`, and with `index_col=0` they are `['Name', 'Age', 'Department', 'Salary']`.

## Advantages of Pandas
- Handles tables with column names, mixed data types and missing values.
- Reads and writes many file formats.
- Very readable code for cleaning, selecting and analysing data.
- Works closely with NumPy, Scikit-learn and plotting libraries.

## Limitations of Pandas
- Needs the whole dataset in memory.

## Interview Questions

**1. What is Pandas and why is it used?**</br>

Pandas is a Python library for working with tabular data. It is used because it can handle tables with column names, different data types and missing values, and it makes loading, cleaning, analysing and saving data easy.

**2. What is the difference between a Series and a DataFrame?**</br>

A Series is a one-dimensional labelled array (one column). A DataFrame is a two-dimensional table with rows and columns, and every column of it is a Series.

**3. What is the difference between a NumPy array and a Pandas DataFrame?**</br>

A NumPy array holds one data type and has no labels, and it is fast for numerical work. A DataFrame has labelled rows and columns, and each column can have its own data type, so it is better for real-world tables.

**4. What is the index in Pandas?**</br>

The index is the set of labels for the rows. By default it is the numbers 0, 1, 2 and so on, but it can be set to something meaningful with `set_index()`. Pandas also uses the index to align data when combining Series or DataFrames.

**5. How do you select one column and several columns?**</br>

`df["col"]` gives one column as a Series, and `df[["col1", "col2"]]` gives several columns as a DataFrame.

**6. How can you create a DataFrame?**</br>

From a dictionary of lists, a list of lists with `columns=`, a list of dictionaries, a NumPy array, a dictionary of Series, or by reading a file with `pd.read_csv()`.

**7. How do you check the data types of the columns, and how do you change them?**</br>

`df.dtypes` shows the types. `astype()` changes the type of a column, and `pd.to_numeric(errors="coerce")` converts text to numbers and turns invalid values into `NaN`.

**8. What does `index=False` do in `to_csv`?**</br>

It stops Pandas from saving the index as an extra first column in the file. Without it, reading the file again creates an extra `Unnamed: 0` column.

**9. What can go wrong if the wrong `sep` is used in `read_csv`?**</br>

The whole line is read as one column, so the table has a single column with the values joined together, and there is no error message. Using the right separator, like `sep=";"`, fixes it.

**10. How do you convert a DataFrame to a NumPy array?**</br>

Use `df.to_numpy()` (or `df.values`). If the DataFrame has only number columns, the array has a number type. If it has mixed types, the array becomes an `object` array.
